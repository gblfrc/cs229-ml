# PS3 - Problem 3 - KL Divergence, Fisher Information, and the Natural Gradient

### (a)

Given the general formula for computing the expected value of a quantity under a given distribution,
$$
E_{y \sim p(y)}\left[g(y) \right] = \int_{-\infty}^{\infty} p(y) g(y) dy
$$

we compute the expected value of the target quantity by replacing such quantity in the previous formula:
$$
E_{y \sim p(y;\theta)}\left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] =  \int_{-\infty}^{\infty} p(y;\theta) \left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] dy
$$

We can now expand the term in squared brackets as:
$$
\nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} = \left. \left[\frac{1}{p(y;\theta')} \nabla_{\theta'} p(y;\theta') \right] \right|_{\theta'=\theta} = \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta)
$$

and update the formula from before accordingly:
$$
\begin{align*}
E_{y \sim p(y;\theta)}\left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] &=  \int_{-\infty}^{\infty} p(y;\theta) \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) dy \\
&=  \int_{-\infty}^{\infty} \nabla_{\theta} p(y;\theta) dy
\end{align*}
$$

At this point, assuming the probability density function at hand is sufficiently regular, the previous equality can be rewritten as:
$$
\begin{align*}
E_{y \sim p(y;\theta)}\left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] &=  \nabla_{\theta} \int_{-\infty}^{\infty}  p(y;\theta) dy \\
&= \nabla_{\theta} 1 \\
&= 0
\end{align*}
$$

where the second equality comes from the fact that the definite integral of a probability distribution function over the whole range of values is always equal to $1$. It follows:
$$ 
E_{y \sim p(y;\theta)}\left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] = 0
$$


### (b)

The covariance matrix of a random vector $X \in \R^n$ is given as:
$$
Cov(X) = E[(X-E[X])(X-E[X])^T]
$$

We adapt the formula to our specific case, replacing for convenience the writing $\nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta}$ with $\nabla_\theta$:
$$
\begin{align*}
\mathcal{I}(\theta) &= Cov_{y \sim p(y;\theta)}\left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] \\
&= E_{y \sim p(y;\theta)}\left[ (\nabla_\theta - E_{y \sim p(y;\theta)}[\nabla_\theta])(\nabla_\theta - E_{y \sim p(y;\theta)}[\nabla_\theta])^T \right]
\end{align*}
$$

The value of the innermost expected values in the formula was found at the previous point to be $0$:
$$
\mathcal{I}(\theta) = E_{y \sim p(y;\theta)}\left[ \nabla_\theta \nabla_\theta^T \right]
$$

Restoring the more complicated writing from before:
$$
\mathcal{I}(\theta) = E_{y \sim p(y;\theta)}\left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta}^T \right]
$$


### (c)

Given the final result from the previous point, the target of the current demonstration can be rewritten as an equality of expected values:
$$
E_{y \sim p(y;\theta)}\left[- \nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta} \right] = E_{y \sim p(y;\theta)}\left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta}^T \right]
$$

We start by computing the Hessian of the starting function, for which each term can be written as:
$$
\begin{align*}
H_{ij} &= \frac{\partial^2 \left[ \log p(y;\theta) \right]}{\partial p(x;\theta)^2} \cdot \frac{\partial \left[ p(y;\theta) \right]}{\partial \theta_i} \cdot \frac{\partial \left[ p(y;\theta) \right]}{\partial \theta_j} + \frac{\partial \left[ \log p(y;\theta) \right]}{\partial p(x;\theta)} \cdot \frac{\partial^2 \left[ p(y;\theta) \right]}{\partial \theta_i \partial \theta_j}\\
&= -\frac{1}{\left(p(x;\theta)\right)^2} \cdot \nabla_{\theta_i} p(y;\theta) \cdot \nabla_{\theta_j} p(y;\theta) + \frac{1}{p(y;\theta)} \cdot \nabla_{\theta_{ij}}^2 p(y;\theta)\\
\end{align*}
$$

It follows immediately that the negative Hessian is:
$$
\begin{align*}
H_{ij} &= \frac{1}{\left(p(x;\theta)\right)^2} \cdot \nabla_{\theta_i} p(y;\theta) \cdot \nabla_{\theta_j} p(y;\theta) - \frac{1}{p(y;\theta)} \cdot \nabla_{\theta_{ij}}^2 p(y;\theta)\\
\end{align*}
$$

and it can be rewritten in vector form as:
$$
- \nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta} = \frac{1}{\left(p(x;\theta)\right)^2} \cdot \nabla_{\theta} p(y;\theta) \cdot \nabla_{\theta}^T p(y;\theta) - \frac{1}{p(y;\theta)} \cdot \nabla_{\theta}^2 p(y;\theta)
$$

We can now compute the expected value of the Hessian, which can be first rewritten as sum of the expected values of the two terms in the right side of the previous equation.
$$
\begin{align*}
E_{y \sim p(y;\theta)}\left[- \nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta} \right] &= E_{y \sim p(y;\theta)}\left[ \frac{1}{\left(p(x;\theta)\right)^2} \cdot \nabla_{\theta} p(y;\theta) \cdot \nabla_{\theta}^T p(y;\theta) - \frac{1}{p(y;\theta)} \cdot \nabla_{\theta}^2 p(y;\theta) \right] \\
&= E_{y \sim p(y;\theta)}\left[ \frac{1}{\left(p(x;\theta)\right)^2} \cdot \nabla_{\theta} p(y;\theta) \cdot \nabla_{\theta}^T p(y;\theta) \right] - E_{y \sim p(y;\theta)} \left[\frac{1}{p(y;\theta)} \cdot \nabla_{\theta}^2 p(y;\theta) \right]
\end{align*}
$$

The second term on the right side of the equation can be expanded following the definition of expected value:
$$
\begin{align*}
E_{y \sim p(y;\theta)} \left[\frac{1}{p} \cdot \nabla_{\theta_{ij}}^2 p(y;\theta) \right] &= \int_{-\infty}^{\infty} p(y;\theta) \frac{1}{p(y;\theta)} \nabla_{\theta}^2 p(y;\theta) dy \\
&= \int_{-\infty}^{\infty} \nabla_{\theta}^2 p(y;\theta) dy \\
&= \nabla_{\theta}^2 \int_{-\infty}^{\infty} p(y;\theta) dy \\
&= \nabla_{\theta}^2 1 \\
&= 0 \\
\end{align*}
$$

where the final equalities are derived following a similar approach as in point **(a)**. The expected value of the Hessian is:
$$
E_{y \sim p(y;\theta)}\left[- \nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta} \right]
= E_{y \sim p(y;\theta)}\left[ \frac{1}{\left(p(x;\theta)\right)^2} \cdot \nabla_{\theta} p(y;\theta) \cdot \nabla_{\theta}^T p(y;\theta) \right]
$$

Moving to the expected value of Fisher Information, we recall from point **(a)** that:
$$
\nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} = \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta)
$$

It follows that: 
$$
\begin{align*}
\nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta}^T &= \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) \frac{1}{p(y;\theta)} \nabla_{\theta}^T p(y;\theta) \\
&= \frac{1}{\left(p(x;\theta)\right)^2} \cdot \nabla_{\theta} p(y;\theta) \cdot \nabla_{\theta}^T p(y;\theta)
\end{align*}
$$

which implies:
$$
E_{y \sim p(y;\theta)}\left[\nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta}^T \right]
= E_{y \sim p(y;\theta)}\left[ \frac{1}{\left(p(x;\theta)\right)^2} \cdot \nabla_{\theta} p(y;\theta) \cdot \nabla_{\theta}^T p(y;\theta) \right]
$$

This final result perfectly matches the expected value of the Hessian, thus demonstrating the target equality.


### (d)

From the definition of the KL divergence:
$$
D_{KL}(p_\theta \| p_{\theta+d}) = \int_{-\infty}^{\infty} p(y;\theta) \log \frac{p(y;\theta)}{p(y;\theta+d)} = \int_{-\infty}^{\infty} p(y;\theta) \left[ \log p(y;\theta) -\log p(y;\theta+d) \right]
$$

Leveraging the Taylor Series expansion presented in the problem description and assuming $f(\tilde{\theta}) = \log p(y;\tilde{\theta})$, we can rewrite the logarithm of the density function as:
$$
\log p(y;\tilde{\theta}) \approx \log p(y;\theta) + (\tilde{\theta}-\theta)^T \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} + \frac{1}{2} (\tilde{\theta}-\theta)^T (\nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta})(\tilde{\theta}-\theta)
$$

By substituting the logarithm approximation in the former integral,
$$
\begin{align*}
D_{KL}(p_\theta \| p_{\theta+d}) &\approx \int_{-\infty}^{\infty} p(y;\theta) \left[ \log p(y;\theta) - \log p(y;\tilde{\theta})|_{\tilde{\theta}=\theta+d} \right] \\
&\approx \int_{-\infty}^{\infty} p(y;\theta) \left[- d^T \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} - \frac{1}{2} d^T (\nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta})d \right] \\
&\approx -\int_{-\infty}^{\infty} p(y;\theta) \left[d^T \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] - \frac{1}{2}\int_{-\infty}^{\infty} p(y;\theta) \left[ d^T (\nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta})d \right] \\
\end{align*}
$$

The 2 integrals in the previous equality can be rewritten as expected values:
$$
\begin{align*}
D_{KL}(p_\theta \| p_{\theta+d}) &\approx -E_{y \sim p(y;\theta)}  \left[d^T \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] - \frac{1}{2}E_{y \sim p(y;\theta)} \left[ d^T (\nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta})d \right] \\
&\approx -d^T E_{y \sim p(y;\theta)}  \left[\nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] + \frac{1}{2} d^T E_{y \sim p(y;\theta)} \left[ -\nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta} \right] d \\
\end{align*}
$$

The 2 expected values in this formula match those computed in the previous points, which were demonstrated to be:
$$
E_{y \sim p(y;\theta)}\left[ \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} \right] = 0
$$
$$
E_{y \sim p(y;\theta)}\left[- \nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta} \right] = \mathcal{I}(\theta)
$$

By substituting these values in the previous formula for KL divergence, we obtain:
$$
\begin{align*}
D_{KL}(p_\theta \| p_{\theta+d}) &\approx \frac{1}{2} d^T \mathcal{I}(\theta) d \\
\end{align*}
$$


### (e)

The constrained optimization problem at hand can be solved following the method of the Lagrange multipliers. Before writing the Lagrangian function, however, we define some useful quantities and highlight the available approximations from either the problem description or the results of the previous points. We consider:
$$
\ell(\theta) = \log p(y;\theta) 
$$
$$
f(d) = \ell(\theta+d) \approx \ell(\theta) + d^T\nabla_{\theta'} \ell(\theta')|_{\theta'=\theta}
$$
$$
g(d) = D_{KL}(p_\theta \| p_{\theta+d}) \approx \frac{1}{2} d^T \mathcal{I}(\theta) d
$$

These quantities can be used to replace the related terms in the generic Lagrangian formula presented in the problem description, thus obtaining:

$$
\begin{align*}
\mathcal{L}(d, \lambda) &= \ell(\theta+d) -\lambda \left[ D_{KL}(p_\theta \| p_{\theta+d}) - c\right] \\
&\approx \ell(\theta) + d^T\nabla_{\theta'} \ell(\theta')|_{\theta'=\theta} - \lambda \left[ \frac{1}{2} d^T \mathcal{I}(\theta) d - c\right] \\
&\approx \log p(y;\theta)  + d^T \nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} - \lambda \left[ \frac{1}{2} d^T \mathcal{I}(\theta) d - c\right] \\
&\approx \log p(y;\theta)  + d^T \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) - \frac{\lambda}{2} d^T \mathcal{I}(\theta) d - \lambda c 
\end{align*}
$$

where in the last equalities we replaced $\nabla_{\theta'} \ell(\theta')|_{\theta'=\theta}$ leveraging the following result from previous points: 
$$
\nabla_{\theta'} \log p(y;\theta')|_{\theta'=\theta} = \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta)
$$

At this point, we compute gradients of the Lagrangian with respect to $d$ and $\lambda$. The optimal result of the problem will be found by setting both gradients to $0$ and solving for $d$ and $\lambda$.
$$
\begin{align*}
\nabla_d \mathcal{L}(d, \lambda) &= \frac{\partial}{\partial d} \left[ \log p(y;\theta)  + d^T \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) - \frac{\lambda}{2} d^T \mathcal{I}(\theta) d - \lambda c  \right] \\
&= \frac{\partial}{\partial d} \left[ d^T \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) - \frac{\lambda}{2} d^T \mathcal{I}(\theta) d \right] \\
&= \frac{1}{p(y;\theta)} \frac{\partial}{\partial d} \left[ d^T  \nabla_{\theta} p(y;\theta)\right] - \frac{\lambda}{2} \frac{\partial}{\partial d} \left[  d^T \mathcal{I}(\theta) d \right] \\
&= \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) - \frac{\lambda}{2} 2 \mathcal{I}(\theta) d \\
&= \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta)  - \lambda \mathcal{I}(\theta) d \\
\end{align*}
$$
$$
\begin{align*}
\nabla_\lambda \mathcal{L}(d, \lambda) &= \frac{\partial}{\partial \lambda} \left[ \log p(y;\theta)  + d^T \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) - \frac{\lambda}{2} d^T \mathcal{I}(\theta) d - \lambda c  \right] \\
&= \frac{\partial}{\partial \lambda} \left[ - \frac{\lambda}{2} d^T \mathcal{I}(\theta) d - \lambda c  \right] \\
&= -\frac{\partial}{\partial \lambda} \left[\lambda \left( \frac{1}{2} d^T \mathcal{I}(\theta) d - c \right) \right] \\
&= c -  \frac{1}{2} d^T \mathcal{I}(\theta) d   \\
\end{align*}
$$

Setting both gradients to 0, we obtain the system of equations:
$$
\begin{cases}
\nabla_d \mathcal{L}(d, \lambda) = 0 \\
\nabla_\lambda \mathcal{L}(d, \lambda) = 0
\end{cases}
\implies
\begin{cases}
\frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta)  - \lambda \mathcal{I}(\theta) d = 0 \\
c - \frac{1}{2} d^T \mathcal{I}(\theta) d = 0
\end{cases}
\implies
\begin{cases}
\frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) = \lambda \mathcal{I}(\theta) d\\
c =  \frac{1}{2} d^T \mathcal{I}(\theta) d
\end{cases}
$$

By solving the first equality in $d$, we obtain:
$$
\tilde{d} = \frac{1}{\lambda p(y;\theta)} \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)
$$

Now we replace $d$ with the just found $\tilde{d}$ value in the second equation and solve for $\lambda$:  

$$
2c = \left(\frac{1}{\lambda p(y;\theta)} \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)\right)^T \mathcal{I}(\theta) \left( \frac{1}{\lambda p(y;\theta)} \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta) \right)
$$
$$
2c = \left(\frac{1}{\lambda p(y;\theta)}\right)^2 \left(\mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)\right)^T \mathcal{I}(\theta) \left(\mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta) \right)
$$
$$
2c = \left(\frac{1}{\lambda p(y;\theta)}\right)^2 \nabla_{\theta}^T p(y;\theta) \mathcal{I}(\theta)^{-T} \mathcal{I}(\theta) \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)
$$
$$
\lambda^2 = \frac{1}{2c\left(p(y;\theta)\right)^2}  \nabla_{\theta}^T p(y;\theta) \mathcal{I}(\theta)^{-T} \mathcal{I}(\theta) \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)
$$
$$
\lambda = \sqrt{\frac{\nabla_{\theta}^T p(y;\theta) \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)}{2c\left(p(y;\theta)\right)^2}}
$$

where in the final equality we used the fact that $\mathcal{I}(\theta)$ is a symmetric covariance matrix and, therefore, $\mathcal{I}(\theta)^{-1} = \mathcal{I}(\theta)^{-T}$. Finally, the value of $\lambda$ just computed can be substituted in the former equation defining $\tilde{d}$ to obtain the final value of the natural gradient $d^*$.

$$
d^* = \lambda^{-1} \frac{1}{p(y;\theta)} \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta) \\
$$
$$
d^* = \sqrt{\frac{2c\left(p(y;\theta)\right)^2}{\nabla_{\theta}^T p(y;\theta) \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)}} \frac{1}{p(y;\theta)} \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta) \\
$$
$$
d^* = \sqrt{\frac{2c}{\nabla_{\theta}^T p(y;\theta) \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)}}  \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta) \\
$$


### (f)

The parameter update rule using Newton's method is:
$$
\theta := \theta - H^{-1} \nabla_\theta \ell(\theta) 
$$

In such formula, the update direction is expressed by the last term on the right side of the equation. Therefore, calling $d_N$ such update direction, it holds:
$$
d_N = - H^{-1} \nabla_\theta \ell(\theta) 
$$

which is the product of the inverse Hessian and the gradient of the log-likelihood function. In the context of the natural gradient method, instead, the update direction is given by the natural gradient itself. For simplicity, we will consider the gradient $\tilde{d}$, which differs from the actual $d^*$ only by a scaling term. From the previous point:
$$
\tilde{d} = \frac{1}{\lambda} \frac{1}{p(y;\theta)} \mathcal{I}(\theta)^{-1} \nabla_{\theta} p(y;\theta)
$$

It has already been shown in the previous points that the gradient of the log-likelihood is:
$$
\nabla_\theta \ell(\theta) = \nabla_{\theta} \log p(y;\theta) = \frac{1}{p(y;\theta)} \nabla_{\theta} p(y;\theta) 
$$

which allows to rewrite $\tilde{d}$ as:
$$
\tilde{d} = \frac{1}{\lambda} \mathcal{I}(\theta)^{-1} \nabla_\theta \ell(\theta)
$$

In addition, it has already been demonstrated that:
$$
\begin{align*}
\mathcal{I}(\theta) &= E_{y \sim p(y;\theta)}\left[- \nabla_{\theta'}^2 \log p(y;\theta')|_{\theta'=\theta} \right] \\
&= E_{y \sim p(y;\theta)}\left[- H \right] \\
&= -E_{y \sim p(y;\theta)}\left[H \right] \\
\end{align*}
$$

Therefore $\tilde{d}$ can be rewritten as:
$$
\tilde{d} = - \frac{1}{\lambda} E_{y \sim p(y;\theta)}\left[H \right]^{-1}  \nabla_\theta \ell(\theta)
$$

which is very similar to the $d_N$ presented at the beginning. Therefore the natural gradient and Newton's method can be said to coincide.
