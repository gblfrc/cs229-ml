### (a)

We start by writing the forward pass of the network at hand for a generic sample $x^{(i)}$. To simplify the notation, we introduce the additional variables $z_j$ and $a$, indicating the intermediate results of the linar combinations before activation. 

$$
z_j^{(i)} = w_{0,j}^{[1]} + w_{1,j}^{[1]} x_1^{(i)} + w_{2,j}^{[1]} x_2^{(i)} 
$$
$$
h_j^{(i)} = \sigma (z_j^{(i)}) 
$$
$$
a^{(i)} = w_{0}^{[2]} + w_{1}^{[2]} h_1^{(i)} + w_{2}^{[2]} h_2^{(i)} + w_{3}^{[2]} h_3^{(i)} \\
$$
$$
o^{(i)} = \sigma(a^{(i)})
$$

With the gradient discend update rule, the target parameter should be updated as:
$$
w_{1,2}^{[1]} := w_{1,2}^{[1]} - \alpha \frac{\partial l}{\partial w_{1,2}^{[1]}}
$$

The gradient term can be expanded and computed as:
$$
\begin{align*}
\frac{\partial l}{\partial w_{1,2}^{[1]}} &= 
\frac{\partial}{\partial w_{1,2}^{[1]}} \left[ \frac{1}{m} \sum_{i=1}^m \big( o^{(i)} - y^{(i)}  \big)^2 \right] \\
&=  \frac{2}{m} \sum_{i=1}^m \big( o^{(i)} - y^{(i)}  \big) \cdot \left[ \frac{\partial}{\partial w_{1,2}^{[1]}}\big( o^{(i)} - y^{(i)}  \big) \right]
\end{align*}
$$

where the term in squared brackets can be expanded using the chain rule.
$$
\frac{\partial}{\partial w_{1,2}^{[1]}}\big( o^{(i)} - y^{(i)}  \big) = 
\frac{\partial o^{(i)}}{\partial  w_{1,2}^{[1]}} = 
\frac{\partial o^{(i)}}{\partial  a^{(i)}}
\frac{\partial a^{(i)}}{\partial  h_2^{(i)}}
\frac{\partial h_2^{(i)}}{\partial z_2^{(i)}}
\frac{\partial z_2^{(i)}}{\partial w_{1,2}^{[1]}}
$$

Now we compute each term singularly:
$$
\frac{\partial o^{(i)}}{\partial a^{(i)}} = \sigma(a^{(i)}) \cdot (1- \sigma(a^{(i)})) = o^{(i)} \cdot (1 - o^{(i)})\\
\frac{\partial a^{(i)}}{\partial  h_2^{(i)}} = w_{2}^{[2]} \\
\frac{\partial h_2^{(i)}}{\partial z_2^{(i)}} = \sigma(z_2^{(i)}) \cdot (1 - \sigma(z_2^{(i)})) = h_2^{(i)} \cdot (1 - h_2^{(i)})\\
\frac{\partial z_2^{(i)}}{\partial w_{1,2}^{[1]}} = x_1^{(i)}
$$

and put the result back into the starting formulae.
$$
\begin{align*}
\frac{\partial l}{\partial w_{1,2}^{[1]}} &= \frac{2}{m} w_{2}^{[2]} \sum_{i=1}^m \big( o^{(i)} - y^{(i)}  \big) o^{(i)} (1 - o^{(i)}) h_2^{(i)} (1 - h_2^{(i)}) x_1^{(i)}
\end{align*}
$$
$$
w_{1,2}^{[1]} := w_{1,2}^{[1]} - \frac{2 \alpha}{m} w_{2}^{[2]} \sum_{i=1}^m \big( o^{(i)} - y^{(i)}  \big) o^{(i)} (1 - o^{(i)}) h_2^{(i)} (1 - h_2^{(i)}) x_1^{(i)}
$$

### (b)

Yes, it is possible. 

We observe that the negative points in the dataset are located within a triangle and that the network is composed by 3 neurons at the first layer. Each of the neurons applies the step function to a linear combination of $x_1$ and $x_2$, which can be used to split the input plane into 2 subplanes with the line:
$$
w_{0,j}^{[1]} + w_{1,j}^{[1]} x_1 + w_{2,j}^{[1]} x_2 = 0
$$

When the term on the left is greater than $0$, the step function would return $1$, which can be used to identify the sub-plane of positive samples. Similarly, the sub-plane where the left term is lower than $0$ identifies the space of negative samples.

The core of the problem is, therefore, to identify the separating lines for the first 3 neurons. From the dataset plot, we observe that two of them are respectively parallel to the $x_2$ and $x_1$ axes and might be:
$$
x_1 = 0.5 \implies  w_{0,1}^{[1]} = -0.5, \quad w_{1,1}^{[1]} = 1, \quad w_{2,1}^{[1]} = 0
$$
$$
x_2 = 0.5 \implies  w_{0,2}^{[1]} = -0.5, \quad w_{1,2}^{[1]} = 0, \quad w_{2,2}^{[1]} = 1
$$
The third line can be the one passing through both points $(0.5, 3.5)$ and $(3.5,0.5)$. Such line is parallel to the bisector of the second and fourth quarter, and is:
$$
x_2 = -x_1 + 4 \implies  w_{0,3}^{[1]} = -4, \quad w_{1,3}^{[1]} = 1, \quad w_{2,3}^{[1]} = 1
$$

As a result, each of the 3 neurons outputs $1$ if the input sample is in its positive subplane. Then, the neurons are aggregated in a second linear combination, to produce the classification score:
$$
o = step(w_{0}^{[2]} + w_{1}^{[2]} h_1 + w_{2}^{[2]} h_2 + w_{3}^{[2]} h_3) \\
$$

The dataset plot shows that a sample, in order to be negative, should lie in the intersection of the 3 negative spaces, which implies that all $h_j$'s must be $0$. Instead, for a sample to be positive, it is necessary that at least one of the $h_j$ terms is $1$. Consequently, the result $o$ should be $0$ when all $h_j$ are $0$ and $1$ otherwise. This can be easily obtained by setting:
$$
w_{1}^{[2]} = 1, \quad w_{2}^{[2]} = 1, \quad w_{3}^{[2]} = 1, \quad and \quad w_{0}^{[2]} = -0.5
$$

This would make the argument of the step function negative only when all $h_j$'s are $0$, and positive otherwise, thus achieving the aforementioned objective.

The coefficient presented in this section were implemented in `src/p01 nn.py`, in the dedicated `optimal step weights` function.


### (c)

No, in the new configuration, it is not possible to find a set of weights providing 100% accuracy on the current dataset.

We first provide an intuitive explanation. With the linear function as activation, the $h_j$ values indicate the distance of the input samples from the lines identified by the neurons. In the former case, it was possible, thanks to the step activation function, to uniquely define on which side of the line the input samples was and the linear combination of the $h_j$'s. allowed to determine whether the point was inside or outside of the triangle. In this case, instead, the
linear combination of the returned distances does not provide univocal information about the point class. It can also be observed that, for a generic horizontal line, the returned distance can be associated both to a poisitive and to a negative sample, thus preventing a successful linear combination.

From a more formal point of view, using the linear function as activation results in making the second layer a direct linear combination of the inputs, thus letting all hidden variables lose their meaning. The forward pass of the network would become:
$$
h_j = w_{0,j}^{[1]} + w_{1,j}^{[1]} x_1 + w_{2,j}^{[1]} x_2
$$
$$
\begin{align*}
a &= w_{0}^{[2]} + \sum_{j=1}^3 w_{j}^{[2]} h_j \\
&= w_{0}^{[2]} + \sum_{j=1}^3 w_{j}^{[2]} \left[ w_{0,j}^{[1]} + w_{1,j}^{[1]} x_1 + w_{2,j}^{[1]} x_2 \right] \\
&= w_{0}^{[2]} + \sum_{j=1}^3 w_{j}^{[2]} w_{0,j}^{[1]} + w_{j}^{[2]} w_{1,j}^{[1]} x_1 + w_{j}^{[2]} w_{2,j}^{[1]} x_2  \\
&= \left[w_{0}^{[2]} + \sum_{j=1}^3 w_{j}^{[2]} w_{0,j}^{[1]}\right] + \left[\sum_{j=1}^3 w_{j}^{[2]} w_{1,j}^{[1]} \right] x_1  + \left[\sum_{j=1}^3 w_{j}^{[2]} w_{2,j}^{[1]} \right] x_2
\end{align*}
$$
$$
o= step(a)
$$

The new shape of $a$ demonstrates that it would become a linear combination of the input variables. It follows that the network would become equivalent to a network with a single neuron, which could work well at separating linearly separable datasets, which is not our case.
