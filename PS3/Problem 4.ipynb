# PS3 - Problem 4 - Semi-supervised EM

### (a)

To demonstrate this assertion, we can follow a similar approach to the one used to demonstrate convergence of the general EM approach on an unsupervised log-likelihood. We start by highlighting that the overall log-likelihood can be expanded as:
$$
\ell_{\text{semi-sup}}(\theta) = \sum_{i=1}^{m} \log \sum_{z^{(i)}} p(x^{(i)}, z^{(i)}; \theta) + \alpha \sum_{i=1}^{\tilde{m}} \log p(\tilde{x}^{(i)}, \tilde{z}^{(i)}; \theta)
$$

in which the first summation on the right is associated to the unsupervised log-likelihood, whereas the second is associated to the supervised one. In the general case of the EM algorithm for an unsupervised setting it was already demonstrated that:
$$
\ell_{\text{unsup}}(\theta) \geq \sum_{i=1}^{m} \sum_{z^{(i)}} Q_i(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)}; \theta)}{Q_i(z^{(i)})} 
$$

Therefore, it holds that:
$$
\ell_{\text{semi-sup}}(\theta) \geq \sum_{i=1}^{m} \sum_{z^{(i)}} Q_i(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)}; \theta)}{Q_i(z^{(i)})}+ \alpha \sum_{i=1}^{\tilde{m}} \log p(\tilde{x}^{(i)}, \tilde{z}^{(i)}; \theta) \quad \forall Q_i, x^{(i)}, \theta
$$

where this inequality holds because the term associated to the supervised log-likelihood is left untouched, whereas only the unsupervised term changes in the way described above. Having identified this writing, it follows that the inequality holds with equality only when the relation for the unsupervised term holds with equality. This justifies the E-step, which is exactly the same as in the general unsupervised setting. Also, we observe that the term on the right of the previous inequality matches with the objective of the maximization problem in the M-step.

We can now prove convergence:
$$
\begin{align*}
\ell_{\text{semi-sup}}(\theta^{(t+1)}) &\geq \sum_{i=1}^{m} \sum_{z^{(i)}} Q_i^{(t)}(z^{(i)})\log \frac{p(x^{(i)}, z^{(i)}; \theta^{(t+1)})}{Q_i^{(t)}(z^{(i)})}+ \alpha \sum_{i=1}^{\tilde{m}} \log p(\tilde{x}^{(i)}, \tilde{z}^{(i)}; \theta^{(t+1)}) \\
&\geq \sum_{i=1}^{m} \sum_{z^{(i)}} Q_i^{(t)}(z^{(i)}) \log \frac{p(x^{(i)}, z^{(i)}; \theta^{(t)})}{Q_i^{(t)}(z^{(i)})}+ \alpha \sum_{i=1}^{\tilde{m}} \log p(\tilde{x}^{(i)}, \tilde{z}^{(i)}; \theta^{(t)}) = \ell_{\text{semi-sup}}(\theta^{(t)})
\end{align*}
$$

where the first inequality stems from the inequality presented above, whereas the second comes from the fact that the first term of the equality is obtained via maximization of the second. The final term coincides with $\ell_{\text{semi-sup}}(\theta^{(t)})$.

### (b)

In the previous point **(a)**, it was already motivated that the only term of the semi-supervised log-likelihood requiring estimation is the one related to the unsupervised portion of the likelihood. As a proof for this, we can observe that in the maximization formula from the previous point, the terms involved in the E-step of the algorithm appear only in the first term on the right, i.e., the one associated to the unsupervised log-likelihood. Therefore, the only elements to estimate in the E-step of this semi-supervised GMM application of EM are the $Q_i(z^{(i)})$ associated to the unsupervised samples. Nothing involving the supervised $\tilde{z}^{(i)}$ terms should be estimated in this step. We can now search for an expression to compute each $Q_i(z^{(i)})$. From the previous point:
$$
Q_i(z^{(i)}) := p(z^{(i)}|x^{(i)};\mu,\Sigma,\phi)
$$

where the generic parameter $\theta$ was replaced by the specific model parameters $\mu$, $\Sigma$, and $\phi$. In the specific context where $z \sim \text{Multinomial}(\phi)$, we can rewrite:
$$
w_j^{(i)} = Q_i(z^{(i)}=j) := p(z^{(i)}=j|x^{(i)};\mu,\Sigma,\phi)
$$

where the $w_j^{(i)}$ was introduced to ease the notation in the following sections. Since the density of $p(z|x)$ is unknown, we should rewrite the previous formula in terms of $p(x|z)$ and $p(z)$:
$$
w_j^{(i)} = \frac{p(x^{(i)}|z^{(i)}=j;\mu,\Sigma) \, p(z^{(i)}=j;\phi)}{\sum_{l=1}^k p(x^{(i)}|z^{(i)}=l;\mu,\Sigma) \, p(z^{(i)}=l;\phi)}
$$

All distributions in this formula are known and their explicit form can be substituted in the expression.
$$
w_j^{(i)} = \frac{\frac{1}{(2\pi)^{d/2} |\Sigma_j|^{1/2}} \exp \left( -\frac{1}{2} (x^{(i)} - \mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) \right) \, \phi_j}
{\sum_{l=1}^k \frac{1}{(2\pi)^{d/2} |\Sigma_l|^{1/2}} \exp \left( -\frac{1}{2} (x^{(i)} - \mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l) \right) \, \phi_l}
$$

### (c)

In the M-step of the algorithm all model parameters must be estimated, namely $\mu$, $\Sigma$, and $\phi$. In order to estimate them, one should compute the gradient of the M-step term presented in point **(a)** with respect to each parameter, set the gradient to 0 and solve for the target parameter. We start by rewriting the M-step term in a generic version involving all the known distributions, to later use as a basis for gradient computation. Adapted from point **(a)**, introducing the knowledge from point **(b)**:
$$
\begin{align*}
\ell_{\text{semi-sup}}(\mu,\Sigma,\phi) &= \sum_{i=1}^{m} \sum_{j=1}^k Q_i(z^{(i)}=j) \log  \frac{p(x^{(i)}, z^{(i)}=j; \mu,\Sigma,\phi)}{Q_i(z^{(i)}=j)} + \alpha \sum_{i=1}^{\tilde{m}} \log p(\tilde{x}^{(i)}, \tilde{z}^{(i)}; \mu,\Sigma,\phi) \\
&= \sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \log \frac{p(x^{(i)}|z^{(i)}=j;\mu,\Sigma) \, p(z^{(i)}=j;\phi)}{w_j^{(i)}} + \alpha \sum_{i=1}^{\tilde{m}} \log \left[p(\tilde{x}^{(i)}|\tilde{z}^{(i)};\mu,\Sigma) \, p(\tilde{z}^{(i)};\phi) \right] \\
&= \sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \log \frac{\frac{1}{(2\pi)^{d/2} |\Sigma_j|^{1/2}} \exp \left( -\frac{1}{2} (x^{(i)} - \mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) \right) \, \phi_j}{w_j^{(i)}} + \\ &\quad\quad \alpha \sum_{i=1}^{\tilde{m}} \log \left[\frac{1}{(2\pi)^{d/2} |\Sigma_{\tilde{z}^{(i)}}|^{1/2}} \exp \left( -\frac{1}{2} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})^T \Sigma_{\tilde{z}^{(i)}}^{-1} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}}) \right) \, \phi_{\tilde{z}^{(i)}} \right] \\
&= \sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \left[ -\log (2\pi)^{d/2} -\log |\Sigma_j|^{1/2} -\frac{1}{2} (x^{(i)} - \mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) +\log \phi_j - \log w_j^{(i)} \right] + \\ &\quad\quad \alpha \sum_{i=1}^{\tilde{m}} \left( - \log (2\pi)^{d/2} -\log |\Sigma_{\tilde{z}^{(i)}}|^{1/2} -\frac{1}{2} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})^T \Sigma_{\tilde{z}^{(i)}}^{-1} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}}) +\log \phi_{\tilde{z}^{(i)}} \right)\\
\end{align*}
$$

Starting from this comprehensive formula, we will compute the gradient with respect to each of the model parameters. We begin with $\mu$:
$$
\begin{align*}
\nabla_{\mu_l} \ell_{\text{semi-sup}}(\mu,\Sigma,\phi)
&= \frac{\partial}{\partial {\mu_l}} \left[\sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \left[ -\log (2\pi)^{d/2} -\log |\Sigma_j|^{1/2} -\frac{1}{2} (x^{(i)} - \mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) +\log \phi_j - \log w_j^{(i)} \right]\right] + \\ &\quad\quad \alpha \frac{\partial}{\partial {\mu_l}} \left[\sum_{i=1}^{\tilde{m}} \left( - \log (2\pi)^{d/2} -\log |\Sigma_{\tilde{z}^{(i)}}|^{1/2} -\frac{1}{2} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})^T \Sigma_{\tilde{z}^{(i)}}^{-1} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}}) +\log \phi_{\tilde{z}^{(i)}} \right) \right]\\
&= -\frac{1}{2} \sum_{i=1}^{m} \sum_{j=1}^k  w_j^{(i)} \frac{\partial}{\partial {\mu_l}} \left[   (x^{(i)} - \mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) \right] -\frac{\alpha }{2} \sum_{i=1}^{\tilde{m}}  \frac{\partial}{\partial {\mu_l}} \left[  (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})^T \Sigma_{\tilde{z}^{(i)}}^{-1} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})  \right]
\end{align*}
$$

To further simplify this expression, we leverage the fact that:
$$
\frac{\partial}{\partial x} \left[x^T A x \right] = 2Ax 
$$

for which a demonstration is provided in the introductory notes about linear algebra. Applied to the gradient in $\mu$, it becomes:
$$
\begin{align*}
\nabla_{\mu_l} \ell_{\text{semi-sup}}(\mu,\Sigma,\phi)
&= -\frac{1}{2} \sum_{i=1}^{m} \sum_{j=1}^k  w_j^{(i)} \frac{\partial}{\partial {\mu_l}} \left[   (x^{(i)} - \mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) \right]-\frac{\alpha }{2} \sum_{i=1}^{\tilde{m}}  \frac{\partial}{\partial {\mu_l}} \left[  (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})^T \Sigma_{\tilde{z}^{(i)}}^{-1} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})  \right] \\
&= -\frac{1}{2} \sum_{i=1}^{m} \sum_{j=1}^k  w_j^{(i)} 2 \Sigma_j^{-1} (x^{(i)} - \mu_j) \frac{\partial}{\partial {\mu_l}} \left[x^{(i)} - \mu_j \right] -\frac{\alpha }{2} \sum_{i=1}^{\tilde{m}}  2\Sigma_{\tilde{z}^{(i)}}^{-1} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}}) \frac{\partial}{\partial {\mu_l}} \left[\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}}  \right] \\
\end{align*}
$$

where the last partial derivative equals $-1$ when $j = l$ or $\tilde{z}^{(i)} = l$, and 0 otherwise. Therefore,
$$
\begin{align*}
\nabla_{\mu_l} \ell_{\text{semi-sup}}(\mu,\Sigma,\phi)
&= \sum_{i=1}^{m}  w_l^{(i)} \Sigma_l^{-1} (x^{(i)} - \mu_l) + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \Sigma_l^{-1} (\tilde{x}^{(i)} - \mu_l) \\
&= \Sigma_l^{-1} \left[ \sum_{i=1}^{m}  w_l^{(i)} (x^{(i)} - \mu_l) + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} (\tilde{x}^{(i)} - \mu_l) \right] \\
&= \Sigma_l^{-1} \left[ \sum_{i=1}^{m}  w_l^{(i)} x^{(i)} - \sum_{i=1}^{m} w_l^{(i)} \mu_l + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \tilde{x}^{(i)} - \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\}\mu_l \right] \\
&= \Sigma_l^{-1} \left[ \sum_{i=1}^{m}  w_l^{(i)} x^{(i)}  + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \tilde{x}^{(i)} - \mu_l \left(\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \right)\right] \\
\end{align*}
$$

The estimator of $\mu_l$ is found by setting the gradient to 0.
$$
\begin{align*}
\Sigma_l^{-1} \left[ \sum_{i=1}^{m}  w_l^{(i)} x^{(i)}  + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \tilde{x}^{(i)} - \mu_l \left(\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \right)\right] &= 0 \\
\sum_{i=1}^{m}  w_l^{(i)} x^{(i)}  + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \tilde{x}^{(i)} - \mu_l \left(\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \right) &= 0
\end{align*}
$$
$$
\mu_l \left(\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \right) = \sum_{i=1}^{m}  w_l^{(i)} x^{(i)}  + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \tilde{x}^{(i)}
$$
$$
\mu_l = \frac{\sum_{i=1}^{m}  w_l^{(i)} x^{(i)}  + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \tilde{x}^{(i)}}{\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\}}
$$

We move on with the computation of gradient in $\Sigma$.
$$
\begin{align*}
\nabla_{\Sigma_l} \ell_{\text{semi-sup}}(\mu,\Sigma,\phi)
&= \frac{\partial}{\partial {\Sigma_l}} \left[\sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \left[ -\log (2\pi)^{d/2} -\log |\Sigma_j|^{1/2} -\frac{1}{2} (x^{(i)} - \mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) +\log \phi_j - \log w_j^{(i)} \right]\right] + \\ &\quad\quad \alpha \frac{\partial}{\partial {\Sigma_l}} \left[\sum_{i=1}^{\tilde{m}} \left( - \log (2\pi)^{d/2} -\log |\Sigma_{\tilde{z}^{(i)}}|^{1/2} -\frac{1}{2} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})^T \Sigma_{\tilde{z}^{(i)}}^{-1} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}}) +\log \phi_{\tilde{z}^{(i)}} \right) \right]\\
&= - \sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \frac{\partial}{\partial {\Sigma_l}}  \left[ \frac{1}{2}\log |\Sigma_j| + \frac{1}{2} (x^{(i)} - \mu_j)^T \Sigma_j^{-1} (x^{(i)} - \mu_j) \right] + 
\\ &\quad\quad - \alpha \sum_{i=1}^{\tilde{m}} \frac{\partial}{\partial {\Sigma_l}} \left[ \frac{1}{2} \log |\Sigma_{\tilde{z}^{(i)}}| + \frac{1}{2} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}})^T \Sigma_{\tilde{z}^{(i)}}^{-1} (\tilde{x}^{(i)} - \mu_{\tilde{z}^{(i)}}) \right]\\
\end{align*}
$$

We can further simplify the previous expression by considering that the content of the partial derivatives return a result different from $0$ only in case the content actually depends on $\Sigma_l$. Similarly to the previous case with the gradient of $\mu_l$, we can rewrite the expression by removing the summation over $j$ in the first term and adding the indicator function in the second.
$$
\begin{align*}
\nabla_{\Sigma_l} \ell_{\text{semi-sup}}(\mu,\Sigma,\phi)
&= - \frac{1}{2} \sum_{i=1}^{m} w_l^{(i)} \frac{\partial}{\partial {\Sigma_l}}  \left[ \log |\Sigma_l| + (x^{(i)} - \mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l) \right] + \\ &\quad\quad - \frac{\alpha}{2} \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\}\frac{\partial}{\partial {\Sigma_l}} \left[  \log |\Sigma_l| + (\tilde{x}^{(i)} - \mu_l)^T \Sigma_l^{-1} (\tilde{x}^{(i)} - \mu_l) \right]\\
&= - \frac{1}{2} \sum_{i=1}^{m} w_l^{(i)} \frac{\partial}{\partial {\Sigma_l}}  \left[ \log |\Sigma_l| + (x^{(i)} - \mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l) \right] + \\ &\quad\quad - \frac{\alpha}{2} \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\}\frac{\partial}{\partial {\Sigma_l}} \left[  \log |\Sigma_l| + (\tilde{x}^{(i)} - \mu_l)^T \Sigma_l^{-1} (\tilde{x}^{(i)} - \mu_l) \right]\\
\end{align*}
$$

We notice that this gradient includes 2 partial derivatives which look completely the same, apart for the difference in the $x^{(i)}$ and $\tilde{x}^{(i)}$ terms. Therefore, we focus on deriving the first one with $x^{(i)}$, and will adjust the second one simply replacing $x^{(i)}$ with $\tilde{x}^{(i)}$. We start by splitting the derivative of the sum of terms into a sum of derivatives, then we will compute each derivative separately.
$$
\frac{\partial}{\partial {\Sigma_l}}  \left[ \log |\Sigma_l| + (x^{(i)} - \mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l) \right] = \frac{\partial}{\partial {\Sigma_l}}  \left[ \log |\Sigma_l| \right] + \frac{\partial}{\partial {\Sigma_l}}  \left[(x^{(i)} - \mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l) \right] 
$$

For the first term, it can be shown that:
$$
\frac{\partial}{\partial {\Sigma_l}}  \left[ \log |\Sigma_l| \right] = \Sigma_l^{-T} = \Sigma_l^{-1}
$$

where in the last equality we leveraged the fact $\Sigma_l$ is a covariance matrix and, as such, it is symmetric. For the second term, we leverage that:
$$
\frac{\partial}{\partial A} \left[x^T A x \right] = x x^T 
$$
$$
\frac{\partial}{\partial A} f(A^{-1}) = -A^{-T} \frac{\partial f(A^{-1})}{\partial A^{-1}} A^{-T}
$$

Therefore,
$$
\begin{align*}
\frac{\partial}{\partial {\Sigma_l}} \left[(x^{(i)} - \mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l) \right] &= -\Sigma_l^{-T} \frac{\partial}{\partial {\Sigma_l^{-1}}} \left[(x^{(i)} - \mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l) \right] \Sigma_l^{-T} \\
&= -\Sigma_l^{-T} (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T \Sigma_l^{-T} \\
&= -\Sigma_l^{-1} (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T \Sigma_l^{-1}
\end{align*}
$$

where in the final equality we leveraged once more the fact that $\Sigma_l$ is symmetric. Back to the starting formula:
$$
\frac{\partial}{\partial {\Sigma_l}}  \left[ \log |\Sigma_l| + (x^{(i)} - \mu_l)^T \Sigma_l^{-1} (x^{(i)} - \mu_l) \right] = \Sigma_l^{-1} - \Sigma_l^{-1} (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T \Sigma_l^{-1}
$$

which can be substituted in the original gradient expression.
$$
\begin{align*}
\nabla_{\Sigma_l} \ell_{\text{semi-sup}}(\mu,\Sigma,\phi)
= &- \frac{1}{2} \sum_{i=1}^{m} w_l^{(i)} \left[ \Sigma_l^{-1} - \Sigma_l^{-1} (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T \Sigma_l^{-1}\right] + \\ &- \frac{\alpha}{2} \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \left[  \Sigma_l^{-1} - \Sigma_l^{-1} (\tilde{x}^{(i)} - \mu_l) (\tilde{x}^{(i)} - \mu_l)^T \Sigma_l^{-1} \right]\\
= &- \frac{1}{2} \sum_{i=1}^{m} w_l^{(i)}  \Sigma_l^{-1} + \frac{1}{2} \sum_{i=1}^{m} w_l^{(i)} \Sigma_l^{-1} (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T \Sigma_l^{-1} + \\ &- \frac{\alpha}{2} \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \Sigma_l^{-1} + \frac{\alpha}{2} \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \Sigma_l^{-1} (\tilde{x}^{(i)} - \mu_l) (\tilde{x}^{(i)} - \mu_l)^T \Sigma_l^{-1}\\
= &- \frac{1}{2} \left(\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \right) \Sigma_l^{-1} \\ &+ \frac{1}{2}\Sigma_l^{-1} \left( \sum_{i=1}^{m} w_l^{(i)}  (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} (\tilde{x}^{(i)} - \mu_l) (\tilde{x}^{(i)} - \mu_l)^T \right) \Sigma_l^{-1}
\end{align*}
$$

The estimator of $\Sigma_l$ can be found by setting the gradient to 0.
$$
\begin{align*}
0 = &- \frac{1}{2} \left(\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \right) \Sigma_l^{-1} \\ &+ \frac{1}{2} \Sigma_l^{-1} \left( \sum_{i=1}^{m} w_l^{(i)}  (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} (\tilde{x}^{(i)} - \mu_l) (\tilde{x}^{(i)} - \mu_l)^T \right) \Sigma_l^{-1}
\end{align*}
$$
$$
\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} = \Sigma_l^{-1} \left( \sum_{i=1}^{m} w_l^{(i)}  (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} (\tilde{x}^{(i)} - \mu_l) (\tilde{x}^{(i)} - \mu_l)^T \right)
$$
$$
\Sigma_l  =  \frac{\sum_{i=1}^{m} w_l^{(i)}  (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} (\tilde{x}^{(i)} - \mu_l) (\tilde{x}^{(i)} - \mu_l)^T }{\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\}} 
$$

Finally, we look for a formula to estimate $\phi$. However, differently from the previous cases, optimization over this parameter should be constrained, since it must hold that $\sum_{j=1}^k \phi_j = 1$. Consequently, the optimal estimator for $\phi$ can be found leveraging the method of the Lagrange multipliers. We start by writing the Lagrangian function:
$$
\mathcal{L}(\phi,\lambda) = \sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \log \phi_j + \alpha \sum_{i=1}^{\tilde{m}} \log \phi_{\tilde{z}^{(i)}} -\lambda \left( \sum_{j=1}^k \phi_j - 1 \right)
$$

where the term on the right was obtained from the generic formula presented above for $\ell_{\text{semi-sup}}(\mu,\Sigma,\phi)$ by removing all terms not related to $\phi$, which would have later returned null derivatives, and adding the constrained optimization term. We can proceed with the derivation of the Lagrangian function with respect to $\phi$ and $\lambda$.
$$
\begin{align*}
\nabla_{\phi_l} \mathcal{L}(\phi,\lambda) &= \frac{\partial}{\partial \phi_l} \left[ \sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \log \phi_j + \alpha \sum_{i=1}^{\tilde{m}} \log \phi_{\tilde{z}^{(i)}} -\lambda \left( \sum_{j=1}^k \phi_j - 1 \right) \right] \\
&= \sum_{i=1}^{m} w_l^{(i)} \frac{1}{\phi_l} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)}=l\} \frac{1}{\phi_l} -\lambda \\
&= \frac{1}{\phi_l} \left(\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)}=l\} \right) -\lambda
\end{align*}
$$
$$
\begin{align*}
\nabla_{\lambda} \mathcal{L}(\phi,\lambda) &= \frac{\partial}{\partial \lambda} \left[ \sum_{i=1}^{m} \sum_{j=1}^k w_j^{(i)} \log \phi_j + \alpha \sum_{i=1}^{\tilde{m}} \log \phi_{\tilde{z}^{(i)}} -\lambda \left( \sum_{j=1}^k \phi_j - 1 \right) \right] \\
&= \sum_{j=1}^k \phi_j - 1
\end{align*}
$$

We can now compute $\phi_l$ and $\lambda$ by setting the gradients to 0. 
$$
\nabla_{\phi_l} \mathcal{L}(\phi,\lambda) = \frac{1}{\phi_l} \left(\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)}=l\} \right) -\lambda = 0
$$
$$
\phi_l = \frac{\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)}=l\}}{\lambda}
$$
$$
\nabla_{\lambda} \mathcal{L}(\phi,\lambda) = \sum_{j=1}^k \phi_j - 1 = 0 
$$
$$
\sum_{j=1}^k \phi_j = 1 \implies \sum_{j=1}^k \frac{\sum_{i=1}^{m} w_j^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)}=j\}}{\lambda} = 1
$$
$$
\begin{align*}
\lambda &= \sum_{i=1}^{m} \sum_{j=1}^k  w_j^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \sum_{j=1}^k \mathbf{1}\{\tilde{z}^{(i)}=j\}  \\
&= \sum_{i=1}^{m} 1 + \alpha \sum_{i=1}^{\tilde{m}} 1 \\
&= m + \alpha \tilde{m}
\end{align*}
$$

where the second equality stems from the fact that the sum of $w_j^{(i)}$ terms is per se a sum of probabilities $Q_i(z^{(i)}=j)$, which must sum to $1$ over all the values of $j$. In conclusion, to recap:
$$
\mu_l = \frac{\sum_{i=1}^{m}  w_l^{(i)} x^{(i)}  + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} \tilde{x}^{(i)}}{\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\}}
$$
$$
\Sigma_l  =  \frac{\sum_{i=1}^{m} w_l^{(i)}  (x^{(i)} - \mu_l) (x^{(i)} - \mu_l)^T + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\} (\tilde{x}^{(i)} - \mu_l) (\tilde{x}^{(i)} - \mu_l)^T }{\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)} = l\}} 
$$
$$
\phi_l = \frac{\sum_{i=1}^{m} w_l^{(i)} + \alpha \sum_{i=1}^{\tilde{m}} \mathbf{1}\{\tilde{z}^{(i)}=l\}}{m + \alpha \tilde{m}}
$$

### (f)

#### Convergence
From running the previous cells, we observe that the semi-supervised case converges much faster than the unsupervised case, requiring on average 35 iterations instead of more than 100, which are required by the unsupervised approach. This result highlights the effectiveness of the supervised samples to accelerate the clustering process even when they are in a significantly lower number with respect to the unlabelled samples (20 vs 980).

#### Stability
For the unsupervised trials, we observe that the color associated to each cluster differs from one plot to another. This implies that, during each trial, each cluster is associated a different number. The different trials all eventually converge to identifying the same clusters, but assigning them a different identifier. On the other hand, the semi-supervised approach always assigns the same number to a specific cluster, due to the fact that the labelled samples are given at the beginning the target cluster number. As a consequence, one can observe that the colors associated to the clusters are always the same and so are the plots.

#### Quality of assignments
The plots show that, in the unsupervised case, the method identifies 2 cluster with a high density of points, and 2 other cluster whose samples are much sparser. On the other hand, the semi-supervised approach identifies 3 high-density clusters and only a single cluster to collect all unhandled samples. The quality of assignments is better in the semi-supervised case, where it matches the expectations given that the data were sampled from 3 low-variance Gaussian distributions and a single high-variance one.
