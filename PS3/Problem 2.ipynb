# PS3 - Problem 2 - KL divergence and Maximum Likelihood

### (a)

We start by rewriting the formula for the KL divergence as:
$$
D_{KL}(P\|Q) = \sum_{x \in \mathcal{X}} P(x) \, \log \frac{P(x)}{Q(x)} = \sum_{x \in \mathcal{X}} P(x) \, \left[-\log \frac{Q(x)}{P(x)} \right]
$$

We can observe that the KL divergence can be written in terms of an expected value:
$$
D_{KL}(P\|Q) = \sum_{x \in \mathcal{X}} P(x) \, \left[-\log \frac{Q(x)}{P(x)} \right] = E_{x \sim P} \left[ -\log \frac{Q(x)}{P(x)} \right] = E_{x \sim P} \left[ f(X) \right]
$$

By computing the derivatives of the negative logarithm function, we can observe that it is a convex function because its second derivative is always greater or equal to $0$:
$$
\begin{align*}
f(x) = -log(x) &\implies f'(x) = -\frac{1}{x} \\
               &\implies f''(x) = \frac{1}{x^2}
\end{align*}
$$

Consequently, from Jensen's inequality:
$$
\begin{align*}
E_{x \sim P} \left[ f(X) \right] &\geq f \left( E_{x \sim P} [X] \right) \\
E_{x \sim P} \left[ -\log \frac{Q(x)}{P(x)} \right] &\geq -\log \left( E_{x \sim P} \left[\frac{Q(x)}{P(x)} \right] \right) \\
D_{KL}(P\|Q) &\geq -\log \left( \sum_{x \in \mathcal{X}} P(x) \, \frac{Q(x)}{P(x)} \right)
\end{align*}
$$

The non-negativity of KL divergence can be demonstrated with some mathematical transformations on the RHS term in the last line:
$$
\begin{align*}
D_{KL}(P\|Q) &\geq -\log \left( \sum_{x \in \mathcal{X}} P(x) \, \frac{Q(x)}{P(x)} \right) \\
&\geq -\log \left( \sum_{x \in \mathcal{X}} Q(x) \right) \\
&\geq -\log 1 \\
&\geq 0 \\
\end{align*}
$$

On top of this, it is easy to demonstrate that $P = Q \implies D_{KL}(P\|Q) = 0$. In such case, 
$$
D_{KL}(P\|Q) = \sum_{x \in \mathcal{X}} P(x) \, \log \frac{P(x)}{Q(x)} = \sum_{x \in \mathcal{X}} P(x) \, \log \frac{P(x)}{P(x)} = \sum_{x \in \mathcal{X}} P(x) \, \log 1 = 0
$$

On the other side, it is less immediate to show that $D_{KL}(P\|Q) = 0 \implies P = Q$. We demonstrate this by observing that, in this case, Jensen's inequality must hold by equality, which is true when the random variable of interest is a constant, i.e.:

$$
\frac{Q(x)}{P(x)} = c
$$

In such case, it also holds that:
$$
E_{x \sim P}\left[\frac{Q(x)}{P(x)} \right] = c
$$

Since Jensen's inequality holds in this case for equality:
$$
\begin{align*}
D_{KL}(P\|Q) &= -\log \left( E_{x \sim P}\left[\frac{Q(x)}{P(x)} \right] \right) =  -\log \left( \sum_{x \in \mathcal{X}} P(x) \, \frac{Q(x)}{P(x)} \right) = -\log \left( \sum_{x \in \mathcal{X}} Q(x) \right) = 0
\end{align*}
$$

from which it follows:
$$
E_{x \sim P}\left[\frac{Q(x)}{P(x)} \right] = 1 \implies c = 1
$$

Consequently:
$$
\frac{Q(x)}{P(x)} = 1 \implies P(x) = Q(x)
$$

which concludes the demonstration. It holds therefore that:
$$
D_{KL}(P\|Q) = 0 \iff P = Q
$$


### (b)

The formula for the KL divergence in a single variable is a summation in which each logarithm term associated to an outcome of the random variable is multiplied by the probability of the term itself. Therefore, its extension to multiple random variables must have a similar form but include the summation on both variables. In particular:
$$
D_{KL}(P(X,Y)\|Q(X,Y)) = \sum_{x} \sum_{y} P(x,y) \, \log \frac{P(x,y)}{Q(x,y)}
$$

We recall that, from the conditional probability definition:
$$
P(y|x) = \frac{P(x,y)}{P(x)} \implies P(x,y) =  P(y|x) P(x)
$$

Consequently, we can expand the previous formula in:
$$
\begin{align*}
D_{KL}(P(X,Y)\|Q(X,Y)) &= \sum_{x} \sum_{y} P(x,y) \, \log \frac{P(x,y)}{Q(x,y)} \\
D_{KL}(P(X,Y)\|Q(X,Y)) &= \sum_{x} \sum_{y} P(y|x) P(x) \, \log \left( \frac{P(y|x)}{Q(y|x)}\cdot \frac{P(x)}{Q(x)} \right) \\
D_{KL}(P(X,Y)\|Q(X,Y)) &= \sum_{x} \sum_{y}  P(y|x) P(x) \, \left( \log \frac{P(y|x)}{Q(y|x)} + \log \frac{P(x)}{Q(x)} \right) \\
D_{KL}(P(X,Y)\|Q(X,Y)) &= \sum_{x} \sum_{y}  P(y|x) P(x) \, \log \frac{P(y|x)}{Q(y|x)} + \sum_{x} \sum_{y}  P(y|x) P(x) \, \log \frac{P(x)}{Q(x)} \\
\end{align*}
$$

The first term on left exactly matches $D_{KL}(P(Y|X)\|Q(Y|X))$. Starting from the provided formula for the KL divergence of the conditional probability of 2 variables:
$$
\begin{align*}
D_{KL}(P(X|Y)\|Q(X|Y)) &= \sum_{y} P(y) \left( \sum_{x} P(x|y) \, \log \frac{P(x|y)}{Q(x|y)} \right) \\
&= \sum_{x} \sum_{y} P(x|y) P(y)\, \log \frac{P(x|y)}{Q(x|y)} \\
\end{align*}
$$

which matches the previously mentioned term, with the only difference of having inverted variables in the conditional probability.

We shift our attention, now, to the second term, and rewrite to re-introduce the joint probability term:
$$
\sum_{x} \sum_{y}  P(y|x) P(x) \, \log \frac{P(x)}{Q(x)} = \sum_{x} \sum_{y} P(x,y) \, \log \frac{P(x)}{Q(x)}
$$

It can be noticed that the only term depending on $y$ is the joint probability term. Therefore:
$$
\sum_{x} \sum_{y} P(x,y) \, \log \frac{P(x)}{Q(x)} = \sum_{x} \log \frac{P(x)}{Q(x)} \sum_{y} P(x,y) 
$$

This formula presents the summation of marginal densities over a specific variable, for which it holds:
$$
\sum_{y} P(x,y) = P(x)
$$

It follows that the previous formula becomes:
$$
\sum_{x} \log \frac{P(x)}{Q(x)} \sum_{y} P(x,y) = \sum_{x} P(x) \log \frac{P(x)}{Q(x)}
$$

where the second term matches the KL divergence for the single variable $x$. Back to the starting relation:
$$
\begin{align*}
D_{KL}(P(X,Y)\|Q(X,Y)) &= \sum_{x} \sum_{y}  P(y|x) P(x) \, \log \frac{P(y|x)}{Q(y|x)} + \sum_{x} P(x) \, \log \frac{P(x)}{Q(x)} \\
D_{KL}(P(X,Y)\|Q(X,Y)) &= D_{KL}(P(Y|X)\|Q(Y|X)) + D_{KL}(P(X)\|Q(X)) \\
\end{align*}
$$



### (c)

We demonstrate that the optimal $\theta$ is the same in both cases by showing that the gradient formulae are the same. The optimal $\theta$ would be found by setting each gradient to $0$. We start with the KL divergence term, which can be written as:
$$
\begin{align*}
D_{KL}(\hat{P}\|P_\theta) &= \sum_{x \in \mathcal{X}} \hat{P}(x) \log \frac{\hat{P}(x)}{P_\theta(x)} \\
D_{KL}(\hat{P}\|P_\theta) &= \sum_{x \in \mathcal{X}} \hat{P}(x) \left( \log \hat{P}(x) - \log P_\theta(x) \right)\\
D_{KL}(\hat{P}\|P_\theta) &= \sum_{x \in \mathcal{X}} \hat{P}(x) \log \hat{P}(x) - \sum_{x \in \mathcal{X}} \hat{P}(x) \log P_\theta(x) \\
\end{align*}
$$

The gradient of this formula with respect to $\theta$ is:
$$
\begin{align*}
\nabla_\theta D_{KL}(\hat{P}\|P_\theta) &= \nabla_\theta \left[\sum_{x \in \mathcal{X}} \hat{P}(x) \log \hat{P}(x) - \sum_{x \in \mathcal{X}} \hat{P}(x) \log P_\theta(x) \right] \\
\nabla_\theta D_{KL}(\hat{P}\|P_\theta) &= \nabla_\theta \left[- \sum_{x \in \mathcal{X}} \hat{P}(x) \log P_\theta(x) \right] \\
\nabla_\theta D_{KL}(\hat{P}\|P_\theta) &= - \sum_{x \in \mathcal{X}} \nabla_\theta \left[\hat{P}(x) \log P_\theta(x) \right] \\
\nabla_\theta D_{KL}(\hat{P}\|P_\theta) &= - \sum_{x \in \mathcal{X}} \hat{P}(x) \frac{1}{P_\theta(x)}\nabla_\theta P_\theta(x) \\
\nabla_\theta D_{KL}(\hat{P}\|P_\theta) &= - \sum_{x \in \mathcal{X}} \left( \frac{1}{m} \sum_{i=1}^m \mathbf{1} \{x^{(i)} = x\} \right)\frac{1}{P_\theta(x)}\nabla_\theta P_\theta(x) \\
\nabla_\theta D_{KL}(\hat{P}\|P_\theta) &= - \frac{1}{m} \sum_{x \in \mathcal{X}} \sum_{i=1}^m \mathbf{1} \{x^{(i)} = x\} \frac{1}{P_\theta(x)}\nabla_\theta P_\theta(x) \\
\end{align*}
$$

For the likelihood term, we observe first that the likelihood function can be rewritten to include a summation over the outcome values.
$$
\begin{align*}
\ell &= \sum_{i=1}^m \log P_\theta(x^{(i)}) \\
&= \sum_{i=1}^m \sum_{x \in \mathcal{X}} \mathbf{1} \{x^{(i)} = x\} \log P_\theta(x) \\
\end{align*}
$$

At this point, we compute its gradient with respect to $\theta$.

$$
\begin{align*}
\nabla_\theta\ell &= \nabla_\theta \left[ \sum_{i=1}^m \sum_{x \in \mathcal{X}} \mathbf{1} \{x^{(i)} = x\} \log P_\theta(x) \right] \\
\nabla_\theta\ell &= \sum_{i=1}^m \sum_{x \in \mathcal{X}} \mathbf{1} \{x^{(i)} = x\}\nabla_\theta \left[ \log P_\theta(x) \right] \\
\nabla_\theta\ell &= \sum_{i=1}^m \sum_{x \in \mathcal{X}} \mathbf{1} \{x^{(i)} = x\} \frac{1}{P_\theta(x)}\nabla_\theta P_\theta(x) \\
\end{align*}
$$

When setting both gradients to 0, we obtain:
$$
\begin{align*}
\nabla_\theta D_{KL}(\hat{P}\|P_\theta) = - \frac{1}{m} \sum_{x \in \mathcal{X}} \sum_{i=1}^m \mathbf{1} \{x^{(i)} = x\} \frac{1}{P_\theta(x)}\nabla_\theta P_\theta(x) &= 0 \\
\sum_{x \in \mathcal{X}} \sum_{i=1}^m \mathbf{1} \{x^{(i)} = x\} \frac{1}{P_\theta(x)}\nabla_\theta P_\theta(x) &= 0 \\
\nabla_\theta\ell = \sum_{i=1}^m \sum_{x \in \mathcal{X}} \mathbf{1} \{x^{(i)} = x\} \frac{1}{P_\theta(x)}\nabla_\theta P_\theta(x) &= 0 \\
\end{align*}
$$

which demonstrates that the optimization objectives perfectly match.
