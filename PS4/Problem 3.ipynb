# PS4 - Problem 3 - PCA

### (a)

From the problem description, we know that $f_u(x)$ is the projection of vector $x$ on the direction of vector $u$, which must lie on the line identified by $u$ for the definition of projection. In addition, in the lecture notes it was already presented that the distance of the projected point from the origin is $x^\top u$. Consequently, since $u$ is a unit vector, the projection can be expressed as:
$$
f_u(x) = \arg \min_{v \in \mathcal{V}} \| x-v \|^2 = x^\top u u
$$

which implies that, for a generic sample $x^{(i)}$ in the dataset:
$$
f_u(x^{(i)}) = (x^{(i)})^\top u u 
$$

It follows that the difference vector of $x^{(i)}$ and $f_u(x^{(i)})$ can be expressed as:
$$
x^{(i)} - f_u(x^{(i)}) = 
\begin{bmatrix}
x^{(i)}_1 - f_u(x^{(i)})_1 \\
x^{(i)}_2 - f_u(x^{(i)})_2 \\
\vdots \\
x^{(i)}_n - f_u(x^{(i)})_n \\
\end{bmatrix}
= 
\begin{bmatrix}
x^{(i)}_1 - (x^{(i)})^\top u u_1 \\
x^{(i)}_2 - (x^{(i)})^\top u u_2 \\
\vdots \\
x^{(i)}_n - (x^{(i)})^\top u u_n \\
\end{bmatrix}
$$

where n is the number of dimensions both of $x^{(i)}$ and of $u$. It follows that the norm of the difference vector is:  
$$
\begin{align*}
\left\| x^{(i)} - f_u(x^{(i)}) \right\|_2^2 &= \sum_{j=1}^n \left(x^{(i)}_j - (x^{(i)})^\top u u_j \right)^2 \\
&= \sum_{j=1}^n \left(x^{(i)}_j\right)^2 + \sum_{j=1}^n \left((x^{(i)})^\top u u_j \right)^2 - 2 \sum_{j=1}^n \left(x^{(i)}_j (x^{(i)})^\top u u_j \right)\\
&= \left\| x^{(i)} \right\| + \left((x^{(i)})^\top u \right)^2 \left\| u \right\|  - 2 \sum_{j=1}^n \sum_{k=1}^n \left(x^{(i)}_j x^{(i)}_k u_k u_j \right)\\
&= \left\| x^{(i)} \right\| + \left((x^{(i)})^\top u \right)^2 \left\| u \right\|  - 2 u^\top x^{(i)} \left(x^{(i)}\right)^\top u \\
\end{align*}
$$

The norm of the difference can be substituted in the target minimization objective:
$$
\begin{align*}
\arg \min_{u:u^\top u = 1} \sum_{i=1}^m \left\| x^{(i)} - f_u(x^{(i)}) \right\|_2^2 &= \arg \min_{u:u^\top u = 1} \sum_{i=1}^m \left[ \left\| x^{(i)} \right\| + \left((x^{(i)})^\top u \right)^2 \left\| u \right\| - 2 u^\top x^{(i)} \left(x^{(i)}\right)^\top u \right] \\
&= \arg \min_{u:u^\top u = 1} \sum_{i=1}^m \left[ - 2 u^\top x^{(i)} \left(x^{(i)}\right)^\top u \right] \\
&= \arg \max_{u:u^\top u = 1} 2 \sum_{i=1}^m \left[ u^\top x^{(i)} \left(x^{(i)}\right)^\top u \right] \\
&= \arg \max_{u:u^\top u = 1} u^\top\left[ \sum_{i=1}^m  x^{(i)} \left(x^{(i)}\right)^\top \right] u \\
\end{align*}
$$

which is the same objective as the one presented in the lecture notes and based on variance maximization. Therefore, the subsequent results must coincide and the $u$ solving the problem is the first principal component.
