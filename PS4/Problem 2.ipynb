# PS4 - Problem 2 - Off Policy Evaluation And Causal Inference

### (a)

From the definition of the expected value over multiple distributions presented in the problem description, we have:
$$
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}}  R(s,a) = \sum_{(s,a)} R(s,a) \, p(s) \,\pi_0(s,a)\\
$$

It follows that, for a more complicated function, it holds:
$$
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[ \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} R(s,a) \right] = \sum_{(s,a)} \left[ \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} R(s,a) \right] p(s) \, \pi_0(s,a)\\
$$

If $\hat{\pi}_0(s,a) = \pi_0(s,a)$, the previous equality can be rewritten as
$$
\begin{align*}
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}}  \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} R(s,a) &= \sum_{(s,a)} \frac{\pi_1(s,a)}{\pi_0(s,a)} R(s,a) \, p(s) \, \pi_0(s,a)\\
&= \sum_{(s,a)}  R(s,a) \, p(s) \, \pi_1(s,a)\\
&= \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} R(s,a)\\
\end{align*}
$$

which matches the demonstration target.
$$
\begin{align*}
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}}  \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} R(s,a)
&= \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} R(s,a)\\
\end{align*}
$$


### (b)

Similarly to the previous case, the expected value at the denominator can be written as:
$$
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}}  \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} = \sum_{(s,a)} \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} \, p(s) \, \pi_0(s,a)
$$

Also, assuming $\hat{\pi}_0(s,a) = \pi_0(s,a)$:
$$
\begin{align*}
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}}  \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} &= \sum_{(s,a)} \frac{\pi_1(s,a)}{\pi_0(s,a)} \, p(s) \, \pi_0(s,a) \\
&= \sum_{(s,a)} \pi_1(s,a) \, p(s) \\
&= \sum_{(s,a)} p(s,a) \\
&= 1 \\
\end{align*}
$$

Therefore, the provided estimator can be rewritten as:
$$
\frac{\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}}  \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} R(s,a)}{\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}}  \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}} = \frac{\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} R(s,a)}{1} = \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} R(s,a)
$$

### (c)

In the case of a finite set of samples, the expected value should be replaced by a summation over the samples. This implies that the estimator can be rewritten as:
$$
\frac{\sum_{i=1}^m \frac{\pi_1(s_i,a_i)}{\hat{\pi}_0(s_i,a_i)} R(s_i,a_i)}{\sum_{i=1}^m \frac{\pi_1(s_i,a_i)}{\hat{\pi}_0(s_i,a_i)}} 
$$

where $m$ is the number of samples in the dataset and $i$ is the index for a generic sample. If the dataset contains a single sample, the estimator can be simplified to:
$$
\frac{\frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} R(s,a)}{\frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}} = R(s,a)
$$

which is the reward from a single observation and does not necessarily match the target of estimation in off-policy evaluation, which is $\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} R(s,a)$.

It follows that, assuming $\pi_0 \neq \pi1$, it holds:
$$
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} R(s,a) \neq \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} R(s,a)
$$

and the estimator is, therefore, biased.

### (d)

Before addressing each of the specific points, we rewrite the starting formula to highlight its building components:
$$
\begin{align*}
&\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[ \mathbb{E}_{a \sim \pi_1(s,a)}  \left[ \hat{R}(s,a) \right] + \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}\left(R(s,a)- \hat{R}(s,a)\right) \right] = \\
&\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[ \mathbb{E}_{a \sim \pi_1(s,a)}  \left( \hat{R}(s,a) \right) \right] + \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[\frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}\left(R(s,a)- \hat{R}(s,a)\right) \right] 
\end{align*}
$$

We can now expand the first component, renaming the variable of the inner expectation $a'$:
$$
\begin{align*}
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[ \mathbb{E}_{a' \sim \pi_1(s,a')} \left( \hat{R}(s,a') \right) \right] &= \sum_{(s,a)} p(s) \, \pi_0(s,a) \, \mathbb{E}_{a' \sim \pi_1(s,a')} \left( \hat{R}(s,a') \right) \\
&= \sum_{(s,a)} p(s) \, \pi_0(s,a) \, \sum_{a'} \hat{R}(s,a')  \pi_1(s,a') \\
&= \sum_s p(s) \, \sum_a \pi_0(s,a) \, \sum_{a'} \hat{R}(s,a')  \pi_1(s,a') \\
&= \sum_s p(s) \, \sum_{a'} \hat{R}(s,a')  \pi_1(s,a') \\
&= \sum_{(s,a')} p(s) \, \hat{R}(s,a')  \pi_1(s,a') \\
&= \mathbb{E}_{\substack{s \sim p(s) \\ a' \sim \pi_1(s,a')}} \left[ \hat{R}(s,a') \right]
\end{align*}
$$

where the third equality stems from the properties of summations and the fourth one follows from the fact that $\sum_a \pi_0(s,a) = 1, \forall s \in S$. By reverting the variable naming, the final result can be written as:
$$
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[ \mathbb{E}_{a' \sim \pi_1(s,a')} \left( \hat{R}(s,a') \right) \right] = \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[ \hat{R}(s,a) \right]
$$

which, substituted in the starting formula, implies:
$$
\begin{align*}
&\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[ \mathbb{E}_{a \sim \pi_1(s,a)}  \left[ \hat{R}(s,a) \right] + \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}\left(R(s,a)- \hat{R}(s,a)\right) \right] = \\
&
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[ \hat{R}(s,a) \right] + \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[\frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}\left(R(s,a)- \hat{R}(s,a)\right) \right] 
\end{align*}
$$

We can now proceed with the 2 specific demonstrations.

#### i

We expand the second term in the estimator formula as done in the previous points:
$$
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[\frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}\left(R(s,a)- \hat{R}(s,a)\right) \right] = \sum_{(s,a)} \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)} \, \left(R(s,a)- \hat{R}(s,a)\right) \, p(s) \, \pi_0(s,a)
$$

Assuming $\hat{\pi}_0(s,a) = \pi_0(s,a)$, the former relation becomes:
$$
\begin{align*}
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[\frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}\left(R(s,a)- \hat{R}(s,a)\right) \right] &= \sum_{(s,a)} \frac{\pi_1(s,a)}{\pi_0(s,a)} \, \left(R(s,a)- \hat{R}(s,a)\right) \, p(s) \, \pi_0(s,a) \\
&= \sum_{(s,a)} \left(R(s,a)- \hat{R}(s,a)\right) \, p(s) \, \pi_1(s,a) \\
&= \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[R(s,a)- \hat{R}(s,a)\right]
\end{align*}
$$

Consequently, the estimator becomes:
$$
\begin{align*}
&\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[ \mathbb{E}_{a \sim \pi_1(s,a)}  \left[ \hat{R}(s,a) \right] + \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}\left(R(s,a)- \hat{R}(s,a)\right) \right] \\
&= \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[ \hat{R}(s,a) \right] + \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[R(s,a)- \hat{R}(s,a)\right] \\
&= \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[ \hat{R}(s,a) \right] + \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[R(s,a) \right] - \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[\hat{R}(s,a)\right] \\
&= \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[R(s,a) \right]\\
\end{align*}
$$

#### ii

For the second case, the result immediately follows from observing that, assuming $R(s,a) = \hat{R}(s,a)$, the second term of the content of the outer expectation becomes null. By replacing $R$ with $\hat{R}$ in the formula, the estimator becomes:
$$
\mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_0(s,a)}} \left[ \mathbb{E}_{a \sim \pi_1(s,a)}  \left[ \hat{R}(s,a) \right] + \frac{\pi_1(s,a)}{\hat{\pi}_0(s,a)}\left(R(s,a)- \hat{R}(s,a)\right) \right] = \mathbb{E}_{\substack{s \sim p(s) \\ a \sim \pi_1(s,a)}} \left[R(s,a) \right]
$$

### (e)

#### i
In this case, the importance sampling estimator is the best solution. Estimating $\hat{R}$ might be difficult, because there is a very complex relation between states, actions and rewards. On the other hand, estimating the probability by which drugs are associated to patients is much easier.

#### ii
In this case, the best option would be the regression estimator, because it allows to estimate the reward function, which is easy to estimate.
