# PS4 - Problem 5 - Markov decision processes

### (a)

From the problem description, we understand that V and B are vectors over the set of states. Therefore, they could be expressed as: 

$$
V_i = 
\begin{bmatrix}
V_i(s_1) \\
V_i(s_2) \\
\vdots \\
V_i(s_n) \\
\end{bmatrix}
\quad\quad\quad
B(V_i) = 
\begin{bmatrix}
R(s_1) + \gamma \max_{a \in A} \sum_{s' \in S} P_{s_1a}(s') V_i(s') \\
R(s_2) + \gamma \max_{a \in A} \sum_{s' \in S} P_{s_2a}(s') V_i(s') \\
\vdots \\
R(s_n) + \gamma \max_{a \in A} \sum_{s' \in S} P_{s_na}(s') V_i(s') \\
\end{bmatrix}
$$

where $i \in \{1,2\}$ for the current problem and $n$ is the cardinality of the finite set of states. It follows that the difference vectors, whose $L_\infty$ norm is considered, are:
$$
V_1 - V_2 = 
\begin{bmatrix}
V_1(s_1) - V_2(s_1) \\
V_1(s_2) - V_2(s_2) \\
\vdots \\
V_1(s_n) - V_2(s_n) \\
\end{bmatrix}
$$
$$
\begin{align*}
B(V_1) - B(V_2) &= 
\begin{bmatrix}
R(s_1) + \gamma \max_{a \in A} \sum_{s' \in S} P_{s_!a}(s') V_1(s') - R(s_1) - \gamma \max_{a \in A} \sum_{s' \in S} P_{s_1a}(s') V_2(s')\\
R(s_2) + \gamma \max_{a \in A} \sum_{s' \in S} P_{s_2a}(s') V_1(s') - R(s_2) - \gamma \max_{a \in A} \sum_{s' \in S} P_{s_2a}(s') V_2(s') \\
\vdots \\
R(s_n) + \gamma \max_{a \in A} \sum_{s' \in S} P_{s_na}(s') V_1(s') - R(s_n) - \gamma \max_{a \in A} \sum_{s' \in S} P_{s_na}(s') V_2(s')\\
\end{bmatrix} \\
&= 
\gamma \begin{bmatrix}
\max_{a \in A} \sum_{s' \in S} P_{s_1a}(s') V_1(s') - \max_{a \in A} \sum_{s' \in S} P_{s_1a}(s') V_2(s')\\
\max_{a \in A} \sum_{s' \in S} P_{s_2a}(s') V_1(s') - \max_{a \in A} \sum_{s' \in S} P_{s_2a}(s') V_2(s') \\
\vdots \\
\max_{a \in A} \sum_{s' \in S} P_{s_na}(s') V_1(s') - \max_{a \in A} \sum_{s' \in S} P_{s_na}(s') V_2(s')\\
\end{bmatrix} \\
&= \gamma D
\end{align*}
$$

where we renamed as $D$ the matrix multiplied by $\gamma$ for convenience. From the definition of $L_\infty$ norm, it follows that:
$$
\begin{align*}
\left\|B(V_1) - B(V_2)\right\|_\infty &= \max_{s \in S} |B(V_1)_s - B(V_2)_s| \\
&= \max_{s \in S} |\gamma D_s| \\
&= \gamma \max_{s \in S} | D_s| \\
&= \gamma \left\|D\right\|_\infty
\end{align*}
$$

Therefore, the target relation to demonstrate becomes:
$$
\begin{align*}
\left\|B(V_1) - B(V_2)\right\|_\infty &\leq \gamma \left\|V_1 - V_2\right\|_\infty \\
\gamma \left\|D\right\|_\infty &\leq \gamma \left\|V_1 - V_2\right\|_\infty \\
\left\|D\right\|_\infty &\leq \left\|V_1 - V_2\right\|_\infty 
\end{align*}
$$

Given the form and contents of $D$, its $L_\infty$ norm can be written as:
$$
\left\|D\right\|_\infty = \max_{s \in S} \left|\max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_1(s') - \max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_2(s') \right|
$$

We can define two relevant quantities:
$$
V_1^M = \max_{s \in S} V_1 \quad\quad\quad
V_2^M = \max_{s \in S} V_2
$$

which are such that:
$$
V_1(s) \leq V_1^M  \quad\quad \text{and} \quad\quad
V_2^M = \max_{s \in S} V_2 \quad\quad\quad \forall s \in S
$$

From this fact and leveraging that that $P_{sa}$ is a probability distribution, i.e., $\sum_{s' \in S} P_{sa}(s') = 1$, we obtain:
$$
\begin{align*}
\left\|D\right\|_\infty &\leq \max_{s \in S} \left|\max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_1^M - \max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_2(s') \right| \\
&\leq \max_{s \in S} \left|\max_{a \in A} V_1^M \sum_{s' \in S} P_{sa}(s')  - \max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_2(s') \right| \\
&\leq \max_{s \in S} \left|V_1^M - \max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_2(s') \right| \\
\end{align*}
$$

For the properties of the absolute value operator, it also follows:
$$
\left|V_1^M - \max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_2(s') \right| = \left|\max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_2(s') - V_1^M \right|
$$

Which allows to further expand the previous inequality:
$$
\begin{align*}
\left\|D\right\|_\infty &\leq \max_{s \in S} \left|\max_{a \in A} \sum_{s' \in S} P_{sa}(s') V_2(s') - V_1^M \right| \\
&\leq \max_{s \in S} \left|V_2^M - V_1^M \right| \\
&\leq \left|V_1^M - V_2^M \right| \\
\end{align*}
$$

At this point, let's assume that $V_1^M = V_1(s)$ for a generic state $s$. The value $ V_2(s)$, associate to the same state $s$ can either be $V_2^M$ or a lower value, from which it follows that: 
$$
\exists \, s \in S : \left|V_1^M - V_2^M \right| \leq \left|V_1(s) - V_2(s) \right|
$$.
This implies that:
$$
\left\|D\right\|_\infty \leq \left|V_1^M - V_2^M \right| \leq \left\|V_1 - V_2\right\|_\infty \implies \left\|D\right\|_\infty \leq \left\|V_1 - V_2\right\|_\infty
$$


### (b)

Let's assume that the Bellman equation admits the existence of at least 2 fixed points:
$$
\exists V_1, V_2 \in \R^n : B(V_1) = V_1  \land  B(V_2) = V_2 
$$

This implies that:
$$
\begin{align*}
\left\|B(V_1) - B(V_2)\right\|_\infty &\leq \gamma \left\|V_1 - V_2\right\|_\infty \\
\left\|V_1 - V_2\right\|_\infty &\leq \gamma \left\|V_1 - V_2\right\|_\infty \\
(1-\gamma)\left\|V_1 - V_2\right\|_\infty &\leq 0 \\
(1-\gamma)\max_{s \in S} \left|V_1(s) - V_2(s) \right| &\leq 0 \\
\end{align*}
$$

We notice that the last inequality includes the maximum of a set of absolute values, which cannot be negative by definition, and the relation can be satisfied only in 2 cases. In the first case, we assume $\max_{s \in S} \left|V_1(s) - V_2(s) \right| > 0$, which implies:
$$
1-\gamma \leq 0 \implies \gamma \geq 1
$$

However, at the beginning of the problem description it was specified that $\gamma < 1$. Therefore the previous case violates an environmental constraint, and cannot be hold valid. To satisfy the inequality, it must hold that:
$$
\max_{s \in S} \left|V_1(s) - V_2(s) \right| = 0
$$

which is satisfied only if:
$$
\left|V_1(s) - V_2(s) \right| = 0 \implies V_1(s) = V_2(s) \quad\quad \forall \, V_1, V_2 \in \R^n
$$

However, this means that $V_1 = V_2$, i.e., the fixed points collapse into a single one and there cannot be 2 different fixed points.
