# PS4 - Problem 4 - Independent components analysis

### (a)

A closed-form solution for $W$ can be found by computing the gradient of the log-likelihood with respect to $W$, setting it to 0, and solving for $W$. Consequently, we start by computing the gradient:

$$
\begin{align*}
\nabla_W \ell(W) &= \nabla_W \left[ \sum_{i=1}^{n}\left(\log|W| + \sum_{j=1}^{d}\log g'\!\left(w_j^{T}x^{(i)}\right)\right) \right] \\
&= \nabla_W \left[ n\log|W| + \sum_{i=1}^{n} \sum_{j=1}^{d}\log g'\!\left(w_j^{T}x^{(i)}\right) \right] \\
&= \nabla_W \left[ n\log|W| \right] + \nabla_W \left[ \sum_{i=1}^{n} \sum_{j=1}^{d}\log g'\!\left(w_j^{T}x^{(i)}\right) \right] \\
&= n W^{-T} + \nabla_W \left[ \sum_{i=1}^{n} \sum_{j=1}^{d}\log g'\!\left(w_j^{T}x^{(i)}\right) \right] \\
\end{align*}
$$

where in the final equality we directly computed the gradient of the first term as  $\nabla_W \left[ \log|W| \right]= W^{-T}$. For the second term, however, it is not as easy to compute the gradient with matrix notation. Therefore, we rewrite $W$ based on its components and express the gradient of a generic $f(W)$ accordingly: 
$$
W = 
\begin{bmatrix}
- \ w_1^T \ - \\
- \ w_2^T \ - \\
\vdots \\
- \ w_d^T \ - \\
\end{bmatrix}
\implies
\nabla_W f(W) = 
\begin{bmatrix}
- \ \nabla_{w_1^T} f(W) \ - \\
- \ \nabla_{w_2^T} f(W) \ - \\
\vdots \\
- \ \nabla_{w_d^T} f(W) \ - \\
\end{bmatrix}
$$

Leveraging this notation, the second term from the previous formula can be expanded as follows, assuming $f(W) = g'\!\left(w_j^{T}x^{(i)}\right)$:
$$
\begin{align*}
\nabla_W \left[ \sum_{i=1}^{n} \sum_{j=1}^{d} \log g'\!\left(w_j^{T}x^{(i)}\right) \right] &= \sum_{i=1}^{n} \sum_{j=1}^{d} \nabla_W \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right] \\
&= \sum_{i=1}^{n} \sum_{j=1}^{d} \begin{bmatrix}
- \ \nabla_{w_1^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right] \ - \\
- \ \nabla_{w_2^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right] \ - \\
\vdots \\
- \ \nabla_{w_d^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right] \ - \\
\end{bmatrix}
\end{align*}
$$

We observe that, in the matrix on the left, each term depends on one component $w_j^{T}$ at a time, which implies that any gradient $\nabla_{w_k^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right]$ would have some value if $j = k$, and be null otherwise, i.e., if $j \neq k$. Consequently, we could include the summation over $j$ in the matrix and write:
$$
\sum_{j=1}^{d} \begin{bmatrix}
- \ \nabla_{w_1^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right] \ - \\
- \ \nabla_{w_2^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right] \ - \\
\vdots \\
- \ \nabla_{w_d^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right] \ - \\
\end{bmatrix}
= 
\begin{bmatrix}
- \ \nabla_{w_1^T} \left[ \log g'\!\left(w_1^{T}x^{(i)}\right) \right] \ - \\
- \ \nabla_{w_2^T} \left[ \log g'\!\left(w_2^{T}x^{(i)}\right) \right] \ - \\
\vdots \\
- \ \nabla_{w_d^T} \left[ \log g'\!\left(w_d^{T}x^{(i)}\right) \right] \ - \\
\end{bmatrix}
$$

This notation shows that each row of the gradient matrix is exactly $\nabla_{w_j^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right]$. The following steps can focus, therefore, on computing such gradient. Since the gradient contains a composition of functions, we firts start by manipulating the derivation objective. From the problem description, we know that $s_j^{(i)} \sim \mathcal{N(0,1)}$ and that $g'(s_j^{(i)})$ is the Gaussisan probability density function on $s_j^{(i)}$. Therefore, we can express $g'$ by substituting the actual values of $\mu$ and $\sigma$ for the distribution. We use the univariate version of the Gaussian density because $s_j^{(i)}$ is a scalar in this problem.
$$
g'\!\left(s_j^{(i)}\right) 
= \frac{1}{\sqrt{2\pi}\sigma} \exp\!\left( -\frac{(s_j^{(i)}-\mu)^2}{2\sigma^2} \right) 
= \frac{1}{\sqrt{2\pi}} \exp\!\left( -\frac{1}{2}\left(s_j^{(i)}\right)^2 \right)
$$

By recalling that $s_j^{(i)} = w_j^{T}x^{(i)}$, the previous formula can be substituted in the gradient computation:
$$
\begin{align*}
\nabla_{w_j^T} \left[ \log g'\!\left(w_j^{T}x^{(i)}\right) \right] 
&= \nabla_{w_j^T} \left[ \log \left[\frac{1}{\sqrt{2\pi}} \exp\!\left( -\frac{1}{2}\left(w_j^{T}x^{(i)}\right)^2 \right)\right] \right] \\
&= \nabla_{w_j^T} \left[ \log \frac{1}{\sqrt{2\pi}} -\frac{1}{2}\left(w_j^{T}x^{(i)}\right)^2 \right] \\
&= -\frac{1}{2} \nabla_{w_j^T} \left[ \left(w_j^{T}x^{(i)}\right)^2 \right] \\
&= - \,w_j^{T}x^{(i)} {x^{(i)}}^T \\
\end{align*}
$$

We can now substitute this expression in the original gradient matrix
$$
\begin{align*}
\nabla_W \left[ \sum_{i=1}^{n} \sum_{j=1}^{d} \log g'\!\left(w_j^{T}x^{(i)}\right) \right] 
&= \sum_{i=1}^{n} \begin{bmatrix}
- \ \nabla_{w_1^T} \left[ \log g'\!\left(w_1^{T}x^{(i)}\right) \right] \ - \\
- \ \nabla_{w_2^T} \left[ \log g'\!\left(w_2^{T}x^{(i)}\right) \right] \ - \\
\vdots \\
- \ \nabla_{w_d^T} \left[ \log g'\!\left(w_d^{T}x^{(i)}\right) \right] \ - \\
\end{bmatrix} \\
&= - \sum_{i=1}^{n} \begin{bmatrix}
- \ w_1^{T}x^{(i)} {x^{(i)}}^T \ - \\
- \ w_2^{T}x^{(i)} {x^{(i)}}^T \ - \\
\vdots \\
- \ w_d^{T}x^{(i)} {x^{(i)}}^T \ - \\
\end{bmatrix} \\
&= - \sum_{i=1}^{n} \begin{bmatrix}
w_1^{T}x^{(i)} \\
w_2^{T}x^{(i)} \\
\vdots \\
w_d^{T}x^{(i)} \\
\end{bmatrix}
{x^{(i)}}^T \\
&= - \sum_{i=1}^{n} W x^{(i)} {x^{(i)}}^T \\
&= - W X^T X
\end{align*}
$$

where the last equalities allows to rewrite the summation over the training samples in terms of the design matrix. The final result can be replaced in the original gradient formula
$$
\nabla_W \ell(W) = n W^{-T} - W X^T X \\
$$

By setting this to 0, one can obtain a formula relating $W$ to the available input design matrix $X$:
$$
\begin{align*}
n W^{-T} - W X^T X &= 0 \\
n {W^T}^{-1} &= W X^T X \\
W^TW &= n (X^T X)^{-1}
\end{align*}
$$

From the final expression, let's assume $\exists \, R \in \R^{d \times d} : R^T R = I$, there exists a generic orthogonal matrix $R$. We observe that, assuming there exists also a matrix $W' = RW$
$$
{W'}^T W' = (RW)^T RW = W^T R^T R W = W^T W
$$

which shows that it is not possible to distinguish a solution $W$ from a solution $W'$. This demonstrates the reason why the ICA approach cannot work assuming Gaussian sources. 

### (b)

The update formula for $W$ is, generally:
$$
W := W + \alpha \nabla_W \ell(W)
$$

where, in this case, $\ell(W)$ is the likelihood associated to a single sample $x^{(i)}$, rather than the sum of likelihood functions over the whole dataset. This means that, adapting the formula presented at the beginning of the previous point:
$$
\ell(W) = \log|W| + \sum_{j=1}^{d}\log g'\!\left(w_j^{T}x^{(i)}\right)
$$

Following the same approach adopted in the previous point, we could initially compute the gradient of the likelihood function as:
$$
\begin{align*}
\nabla_W \ell(W) &= \nabla_W \left[ \log|W| + \sum_{j=1}^{d}\log g'\!\left(w_j^{T}x^{(i)}\right) \right] \\
&= \nabla_W \left[ \log|W| + \sum_{j=1}^{d}\log g'\!\left(w_j^{T}x^{(i)}\right) \right] \\
&= W^{-T} + \nabla_W \left[ \sum_{j=1}^{d}\log g'\!\left(w_j^{T}x^{(i)}\right) \right] \\
\end{align*}
$$

From the problem description, we know that $s_j^{(i)} \sim \mathcal{L(0,1)}$, and that the probability density function for such a random variable is $f_{\mathcal{L}}(s) = \frac{1}{2} \exp(-|s|)$. Given the structure of the likelihood function, in this context $f_{\mathcal{L}}(s) = g'(s)$, which implies:
$$
g'(w_j^{T}x^{(i)}) = \frac{1}{2} \exp(-|w_j^{T}x^{(i)}|) 
$$

It follows that:
$$
\begin{align*}
\nabla_W \ell(W) &= W^{-T} + \nabla_W \left[ \sum_{j=1}^{d}\log \left[\frac{1}{2} \exp(-|w_j^{T}x^{(i)}|) \right] \right] \\
&= W^{-T} + \sum_{j=1}^{d} \nabla_W \left[ \log \frac{1}{2} -|w_j^{T}x^{(i)}|  \right]  \\
&= W^{-T} + \sum_{j=1}^{d} \nabla_W \left[ -|w_j^{T}x^{(i)}|  \right]  \\
\end{align*}
$$

The second term on the right-hand side of the equality can be expanded in a similar way as in the previous point:
$$
\sum_{j=1}^{d} \nabla_W \left[ -|w_j^{T}x^{(i)}|  \right] =
\sum_{j=1}^{d} \begin{bmatrix}
- \ \nabla_{w_1^T} \left[ -|w_j^{T}x^{(i)}|  \right] \ - \\
- \ \nabla_{w_2^T} \left[ -|w_j^{T}x^{(i)}|  \right] \ - \\
\vdots \\
- \ \nabla_{w_d^T} \left[ -|w_j^{T}x^{(i)}|  \right] \ - \\
\end{bmatrix}
= 
\begin{bmatrix}
- \ \nabla_{w_1^T} \left[ -|w_1^{T}x^{(i)}|  \right] \ - \\
- \ \nabla_{w_2^T} \left[ -|w_2^{T}x^{(i)}|  \right] \ - \\
\vdots \\
- \ \nabla_{w_d^T} \left[ -|w_d^{T}x^{(i)}|  \right] \ - \\
\end{bmatrix}
$$

Again, similarly to the previous case, each line in the gradient matrix can be computed per se, with the general formula
$$
\nabla_{w_j^T} \left[-|w_j^{T}x^{(i)}| \right] = - \nabla_{w_j^T} \left[|w_j^{T}x^{(i)}|  \right] = -\text{sign}(w_j^{T}x^{(i)})  {x^{(i)}}^T   \\
$$

The result of this computation can be replaced in the gradient matrix, thus obtaining:
$$
-
\begin{bmatrix}
- \ \text{sign}(w_1^{T}x^{(i)})  {x^{(i)}}^T \ - \\
- \ \text{sign}(w_2^{T}x^{(i)})  {x^{(i)}}^T \ - \\
\vdots \\
- \ \text{sign}(w_d^{T}x^{(i)})  {x^{(i)}}^T \ - \\
\end{bmatrix}
= 
-
\begin{bmatrix}
- \ \text{sign}(w_1^{T}x^{(i)}) \ - \\
- \ \text{sign}(w_2^{T}x^{(i)}) \ - \\
\vdots \\
- \ \text{sign}(w_d^{T}x^{(i)}) \ - \\
\end{bmatrix}
{x^{(i)}}^T
= 
- \text{sign} \left(
\begin{bmatrix}
- \ w_1^{T}x^{(i)} \ - \\
- \ w_2^{T}x^{(i)} \ - \\
\vdots \\
- \ w_d^{T}x^{(i)} \ - \\
\end{bmatrix} \right)
{x^{(i)}}^T
= 
- \text{sign} \left( W x^{(i)} \right) {x^{(i)}}^T
$$

where in the last 2 equalities, we introduced the notation for which applying the $\text{sign}()$ function to a vector returns the vector of the results of the function on each element of the starting vector. The gradient formula becomes, therefore:
$$
\nabla_W \ell(W) = W^{-T} - \text{sign} \left( W x^{(i)} \right) {x^{(i)}}^T  \\
$$

Consequently, the update formula for $W$ becomes
$$
W := W + \alpha \left( W^{-T} - \text{sign} \left( W x^{(i)} \right) {x^{(i)}}^T \right)
$$
