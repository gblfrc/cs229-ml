# PS1 - Problem 4 - Convexity of Generalized Linear Models

### (a)

From the properties of the probability density functions, we know that:

$$
\int_{-\infty}^{\infty} p(y; \eta)\, dy = 1 \quad \forall \eta \in \R
$$

This formula describes the equality between 2 functions of $\eta$, namely the integral of the probability density and the $1$ function. If such condition holds, then also the derivatives of both functions must always be equal:

$$
\frac{\partial}{\partial \eta} \int_{-\infty}^{\infty} p(y; \eta)\, dy = \frac{\partial}{\partial \eta}1 = 0
$$

For the properties of integrals:

$$
\frac{\partial}{\partial \eta} \int p(y; \eta)\, dy
=
\int \frac{\partial}{\partial \eta} p(y; \eta)\, dy
$$

where the ${-\infty}$ and ${\infty}$ boundaries of the integral have been removed only for convenience. The integrals in the following parts are still definite integrals over the whole support. Now, we can compute the partial derivative on $\eta$ in the integral. 

$$
\begin{align*}
\int \frac{\partial}{\partial \eta} p(y; \eta)\, dy &= \int \frac{\partial}{\partial \eta} b(y) \exp (\eta y - a (\eta)) \, dy \\
&= \int b(y) \exp (\eta y - a (\eta)) \frac{\partial}{\partial \eta} \left[\eta y - a (\eta)\right] \, dy \\
&= \int b(y) \exp (\eta y - a (\eta)) \left( y - \frac{\partial a (\eta)}{\partial \eta}\right) \, dy \\
&= \int y p(y; \eta) - \frac{\partial a (\eta)}{\partial \eta} p(y; \eta)\, dy \\
\end{align*}
$$

Leveraging the properties of integrals and the definition of expectation, the formula can be rewritten as:
$$
\begin{align*}
\int \frac{\partial}{\partial \eta} p(y; \eta)\, dy
&= \int y p(y; \eta) dy - \int\frac{\partial a (\eta)}{\partial \eta} p(y; \eta)\, dy \\
&= E[y;\eta] - \frac{\partial a (\eta)}{\partial \eta} \int p(y; \eta)\, dy \\
&= E[y;\eta] - \frac{\partial a (\eta)}{\partial \eta}
\end{align*}
$$

Returning to the starting equality:
$$
\frac{\partial}{\partial \eta} \int p(y; \eta)\, dy = E[y;\eta] - \frac{\partial a (\eta)}{\partial \eta} = 0
$$

from which it follows immediately:
$$
E[y;\eta] = \frac{\partial a (\eta)}{\partial \eta} \implies E[y|x;\theta] = \frac{\partial a (\eta)}{\partial \eta}, \quad with \ \ \eta = \theta^T x
$$


### (b)

The point above allows to write that: 
$$
E[y;\eta] = \int y p(y; \eta)\, dy = \frac{\partial a (\eta)}{\partial \eta}
$$

Following a similar reasoning to the one used above, if the two functions specified in the last equation are equal, so are their derivatives. Therefore,
$$
\frac{\partial}{\partial \eta}\int y p(y; \eta)\, dy = \frac{\partial^2 a (\eta)}{\partial \eta^2}
$$

Now, one can expand the derivative on the left:
$$
\begin{align*}
\frac{\partial}{\partial \eta}\int y p(y; \eta)\, dy &= \int y \frac{\partial}{\partial \eta} \left[ p(y; \eta)\right]\, dy \\
&= \int y \frac{\partial}{\partial \eta} \left[ b(y) \exp (\eta y - a (\eta)) \right]\, dy \\
&= \int y b(y) \exp (\eta y - a (\eta)) \left( y - \frac{\partial a (\eta)}{\partial \eta}\right)\, dy \\
&= \int y p(y; \eta) \left( y - \frac{\partial a (\eta)}{\partial \eta}\right)\, dy \\
&= \int y^2 p(y; \eta) - \frac{\partial a (\eta)}{\partial \eta}y p(y; \eta)\, dy \\
&= \int y^2 p(y; \eta) - \int \frac{\partial a (\eta)}{\partial \eta}y p(y; \eta)\, dy \\
&= \int y^2 p(y; \eta) - \frac{\partial a (\eta)}{\partial \eta} \int y p(y; \eta)\, dy \\
&= \int y^2 p(y; \eta) - \frac{\partial a (\eta)}{\partial \eta} \int y p(y; \eta)\, dy \\
\end{align*}
$$

All terms in the last equation can be rewritten as expectations of $y$: 
$$
\begin{align*}
\frac{\partial}{\partial \eta}\int y p(y; \eta)\, dy &= E[y^2; \eta] - E[y;\eta] E[y;\eta] \\
&= E[y^2; \eta] - E[y;\eta]^2 = Var[y; \eta]\\
\end{align*}
$$

Therefore, the original equality is:
$$
Var[y; \eta] = \frac{\partial^2 a (\eta)}{\partial \eta^2} \implies Var[y|x; \theta] = \frac{\partial^2 a (\eta)}{\partial \eta^2}, \quad with \ \ \eta = \theta^T x
$$


### (c)

Given the likelihood function 
$$
L(\theta) = \prod_{i=1}^m p(y^{(i)}|x^{(i)};\theta)
$$

the negative log-likelihood (NLL) function has the form:
$$
\ell(\theta) = - \sum_{i=1}^m \log p(y^{(i)}|x^{(i)};\theta) 
$$

We can expand the probability term to have a better formulation of the NLL:
the negative log-likelihood (NLL) function has the form:
$$
\begin{align*}
\ell(\theta) &= - \sum_{i=1}^m \log \left[b(y^{(i)}) \exp (\theta^T x^{(i)} y^{(i)} - a (\theta^T x^{(i)})) \right] \\
&= - \sum_{i=1}^m \log b(y^{(i)}) + \theta^T x^{(i)} y^{(i)} - a (\theta^T x^{(i)}) \\
&= \sum_{i=1}^m a (\theta^T x^{(i)}) - \log b(y^{(i)}) - \theta^T x^{(i)} y^{(i)} 
\end{align*}
$$

To compute the Hessian, we start searching for the formulation of each of its terms.
$$
H_{jk} = \frac{\partial^2}{\partial \theta_j \partial \theta_k} \left[ \sum_{i=1}^m a (\theta^T x^{(i)}) - \log b(y^{(i)}) - \theta^T x^{(i)} y^{(i)} \right]
$$

For the sake of simplicity, since the derivative of a summation equals the summation of derivatives, we will compute it considering at first a single sample. We will re-introduce the summation later.

$$
\frac{\partial^2}{\partial \theta_j \partial \theta_k} \left[ a (\theta^T x^{(i)}) - \log b(y^{(i)}) - \theta^T x^{(i)} y^{(i)} \right] = \frac{\partial^2}{\partial \theta_j \partial \theta_k}  a (\theta^T x^{(i)}) - \frac{\partial^2}{\partial \theta_j \partial \theta_k} \theta^T x^{(i)} y^{(i)}
$$

Now we can expand each term on the right of the equality:
$$
\begin{align*}
\frac{\partial^2}{\partial \theta_j \partial \theta_k}  a (\theta^T x^{(i)})  &= 
\frac{\partial^2 a (\theta^T x^{(i)})}{\partial \theta^T x^{(i)}}  \frac{\partial \left(\theta^T x^{(i)}\right)}{\partial \theta_j} \frac{\partial \left(\theta^T x^{(i)}\right)}{\partial \theta_k} \\
&= Var[y|x;\theta] x^{(i)}_j x^{(i)}_k 
\end{align*}
$$

$$
\begin{align*}
\frac{\partial^2}{\partial \theta_j \partial \theta_k} \theta^T x^{(i)} y^{(i)}  &= 
\frac{\partial^2 (\theta^T x^{(i)} y^{(i)})}{\partial \theta^T x^{(i)}}  \frac{\partial \left(\theta^T x^{(i)}\right)}{\partial \theta_j} \frac{\partial \left(\theta^T x^{(i)}\right)}{\partial \theta_k} \\
&= 0 \, x^{(i)}_j x^{(i)}_k = 0
\end{align*}
$$

Therefore, after re-introducing the summation, 
$$
H_{jk} = \sum_{i=1}^m Var[y|x;\theta] x^{(i)}_j x^{(i)}_k
$$

To verify that the Hessian is PSD, we should compute its quadratic form and verify that it is always greater or equal to zero.

$$
\begin{align*}
z^T H z &= \sum_{j=1}^n \sum_{k=1}^n \sum_{i=1}^m Var[y|x;\theta] x^{(i)}_j x^{(i)}_k x^{(i)}_j x^{(i)}_k \\
&= \sum_{j=1}^n \sum_{k=1}^n \sum_{i=1}^m Var[y|x;\theta] \left( x^{(i)}_j x^{(i)}_k \right)^2
\end{align*}
$$

Since $Var[y|x;\theta] \geq 0$, it follows that also $z^T H z \geq 0$.
