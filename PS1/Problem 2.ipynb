### (a)

The target formula can be derived from the joint probability of $y^{(i)} = 1$, $t^{(i)} = 1$ and x^{(i)}, using the chain rule to expand the probability in different ways.

$$
p\left(y^{(i)} = 1, t^{(i)} = 1, x^{(i)}\right) = p\left(y^{(i)} = 1 \mid t^{(i)} = 1, x^{(i)}\right) \ p\left(t^{(i)} = 1, x^{(i)}\right)
$$

$$
p\left(y^{(i)} = 1, t^{(i)} = 1, x^{(i)}\right) = p\left(y^{(i)} = 1 \mid t^{(i)} = 1, x^{(i)}\right) \ p\left(t^{(i)} = 1 \mid x^{(i)}\right) \ p\left(x^{(i)}\right)
$$

$$
p\left(t^{(i)} = 1, y^{(i)} = 1, x^{(i)}\right) = p\left(y^{(i)} = 1 \mid t^{(i)} = 1, x^{(i)}\right) \ p\left(t^{(i)} = 1 \mid x^{(i)}\right) \ p\left(x^{(i)}\right)
$$

$$
p\left(t^{(i)} = 1 \mid y^{(i)} = 1, x^{(i)}\right) p\left(y^{(i)} = 1, x^{(i)}\right) = p\left(y^{(i)} = 1 \mid t^{(i)} = 1, x^{(i)}\right) \ p\left(t^{(i)} = 1 \mid x^{(i)}\right) \ p\left(x^{(i)}\right)
$$

$$
p\left(t^{(i)} = 1 \mid y^{(i)} = 1, x^{(i)}\right) p\left(y^{(i)} = 1 \mid x^{(i)}\right) p\left(x^{(i)}\right) = p\left(y^{(i)} = 1 \mid t^{(i)} = 1, x^{(i)}\right) \ p\left(t^{(i)} = 1 \mid x^{(i)}\right) \ p\left(x^{(i)}\right)
$$

$$
p\left(t^{(i)} = 1 \mid y^{(i)} = 1, x^{(i)}\right) p\left(y^{(i)} = 1 \mid x^{(i)}\right) = p\left(y^{(i)} = 1 \mid t^{(i)} = 1\right) \ p\left(t^{(i)} = 1 \mid x^{(i)}\right)
$$

$$
\frac{p\left(t^{(i)} = 1 \mid y^{(i)} = 1, x^{(i)}\right)}{p\left(y^{(i)} = 1 \mid t^{(i)} = 1\right)} p\left(y^{(i)} = 1 \mid x^{(i)}\right) =  \ p\left(t^{(i)} = 1 \mid x^{(i)}\right)
$$

The numerator of the first fraction can be further simplified. In particular, from the definition of conditional probability, it can be written as:

$$
p\left(t^{(i)} = 1 \mid y^{(i)} = 1, x^{(i)}\right) = \frac{p\left(t^{(i)} = 1, y^{(i)} = 1, x^{(i)}\right)}{p\left(y^{(i)} = 1, x^{(i)}\right)}
$$

From the same definition of conditional probability, and given the problem constraints:

$$
p\left(t^{(i)} = 1 \mid y^{(i)} = 1 \right) = \frac{p\left(t^{(i)} = 1, y^{(i)} = 1\right)}{p\left(y^{(i)} = 1\right)} = 1
$$

$$
p\left(t^{(i)} = 1, y^{(i)} = 1\right) = p\left(y^{(i)} = 1\right)
$$

which means that the intersection of $y^{(i)} = 1$ and $t^{(i)} = 1$ coincides with $y^{(i)} = 1$ itself. This implies:

$$
p\left(t^{(i)} = 1, y^{(i)} = 1, x^{(i)}\right) = {p\left(y^{(i)} = 1, x^{(i)}\right)} \implies \left(t^{(i)} = 1 \mid y^{(i)} = 1, x^{(i)}\right) = \frac{p\left(y^{(i)} = 1, x^{(i)}\right)}{p\left(y^{(i)} = 1, x^{(i)}\right)} = 1
$$

Therefore, the previous formula becomes:

$$
\frac{1}{p\left(y^{(i)} = 1 \mid t^{(i)} = 1\right)} p\left(y^{(i)} = 1 \mid x^{(i)}\right) =  \ p\left(t^{(i)} = 1 \mid x^{(i)}\right)
$$

thus matching the target formula with:

$$
\alpha = p\left(y^{(i)} = 1 \mid t^{(i)} = 1\right)
$$

### (b)

In the previous point, it was demonstrated that:

$$
\frac{p\left(y^{(i)} = 1 \mid x^{(i)}\right)}{\alpha}  =  \ p\left(t^{(i)} = 1 \mid x^{(i)}\right)
$$

Adding the information from the problem description, the formula above becomes:

$$
\frac{h\left(x^{(i)}\right)}{\alpha} \approx \frac{p\left(y^{(i)} = 1 \mid x^{(i)}\right)}{\alpha}  =  \ p\left(t^{(i)} = 1 \mid x^{(i)}\right) \approx 1 \quad\quad \forall x^{(i)} \in V_+
$$

This implies that, $\forall x^{(i)} \in V_+$

$$
\frac{h\left(x^{(i)}\right)}{\alpha} \approx 1 \implies h\left(x^{(i)}\right) \approx \alpha
$$

which is the aim of the demonstration.

\exp (-\theta^T x) = \frac{2}{\alpha} - 1
$$
$$
-\theta^T x = \ln \left( \frac{2}{\alpha} - 1 \right)
$$
$$
\theta^T x + \ln \left( \frac{2}{\alpha} - 1 \right) = 0
$$
$$
\theta_2 x_2 + \theta_1 x_1 + \theta_0 + \ln \left( \frac{2}{\alpha} - 1 \right) = 0
