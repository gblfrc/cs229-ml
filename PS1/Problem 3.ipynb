# PS1 - Problem 3 - Poisson Regression

### (a)

To belong to the exponential family, the starting Poisson distribution
$$
p(y;\lambda) = \frac{e^{-\lambda} \lambda^y}{y!}
$$

should present all terms related to $\lambda$ at the exponent of $e$. Therefore, start by putting the whole formula as exponent of $e$, and adjust it with support of $\log$ functions.

$$
\begin{align*}
p(y;\lambda) &= \exp \left( \log \left( \frac{\lambda^y}{y!} e^{-\lambda} \right) \right) = \exp \left( \log (\lambda^y) + \log (e^{-\lambda}) -\log (y!) \right) \\
&= \exp \left( y\log \lambda -\lambda\log e + \log (\frac{1}{y!}) \right) \\
&= \frac{1}{y!} \exp \left( y\log \lambda -\lambda \right) \\
\end{align*}

$$

Given the previous formula and the form of exponential family distributions
$$
p(y;\eta) = b(y) \exp (\eta^T T(y) - a (\eta)),
$$

the Poisson distribution is shown to belong to such family assuming:
$$
\begin{cases}
    b(y) = \frac{1}{y!} \\
    \eta = \log \lambda \\
    T(y) = y \\
    a (\eta) = e^\eta
\end{cases}

$$

### (b)

The canonical distribution function is a $g(\eta)$ function, such that:
$$
g(\eta) = E \left[T(y); \eta \right].
$$

From the previous point, we observe that in this case $T(y) = y$, therefore:
$$
g(\eta) = E \left[y; \eta \right].
$$

In a Poisson distribution, such as y, the expected value above coincides with $\lambda$. Therefore:
$$
g(\eta) = \lambda = e^\eta.
$$

In the context of Poisson regression with a GLM, the reasoning above becomes:
$$
g(\theta) = E \left[y|x; \theta \right] = e^{\theta^T x},
$$

assuming $\eta = \theta^T x$, which holds in general for GLMs. The above formula coincides with the hypothesis $h_\theta(x)$.

### (c)

We know that the generic formula for gradient ascent is

$$
\theta := \theta + \alpha \nabla_\theta \ell(\theta)
$$

which requires the computation of the log-likelihood gradient to be fully expanded. We start by expressing the likelihood and log-likelihood of the current Poisson distribution, assuming all samples are independent and identically distributed.

$$
L(\theta) = \prod_{i=1}^{m} p(y|x;\theta) = \prod_{i=1}^{m} \frac{1}{y^{(i)}!} \exp \left( y^{(i)}{\theta^T x^{(i)}} -e^{\theta^T x^{(i)}} \right)
$$
$$
\ell(\theta) = \sum_{i=1}^{m} \log \left[ \frac{1}{y^{(i)}!} \exp \left( y^{(i)}{\theta^T x^{(i)}} -e^{\theta^T x^{(i)}} \right) \right] = \sum_{i=1}^{m} \left[ \log \left( \frac{1}{y^{(i)}!}\right) + y^{(i)}{\theta^T x^{(i)}} -e^{\theta^T x^{(i)}} \right]
$$

Now, one can compute the gradient of the log-likelihood. Since the gradient of a sum is the sum of gradients, we initially simplify the notation by computing the gradient over a single sample.

$$
\begin{align*}
\nabla_\theta \ell(\theta) &= \frac{\partial}{\partial \theta} \left[ \log \left( \frac{1}{y^{(i)}!}\right) + y^{(i)}{\theta^T x^{(i)}} -e^{\theta^T x^{(i)}} \right] = \frac{\partial}{\partial \theta} \left[  y^{(i)}{\theta^T x^{(i)}} -e^{\theta^T x^{(i)}} \right] \\

&=  \frac{\partial}{\partial \theta} \left( y^{(i)}{\theta^T x^{(i)}}\right) -  \frac{\partial}{\partial \theta} \left( e^{\theta^T x^{(i)}} \right) \\

&=  y^{(i)}x^{(i)} -  e^{\theta^T x^{(i)}}x^{(i)}  = \left( y^{(i)} -  e^{\theta^T x^{(i)}} \right) x^{(i)}
\end{align*} 
$$

Therefore, the formula for stochastic gradient ascent would become:
$$
\begin{align*}
\theta &:= \theta + \alpha \left( y^{(i)} - e^{\theta^T x^{(i)}} \right) x^{(i)} \\
\theta &:= \theta + \alpha \left( y^{(i)} - h_\theta \left(x^{(i)} \right) \right) x^{(i)}.
\end{align*}
$$

As mentioned before, the summation of terms in the log-likelihood was neglected for simplicity. Computing the overall gradient of the summation would lead to a generic formula for batch gradient ascent.

$$
\begin{align*}
\nabla_\theta \ell(\theta) &= \sum_{i=1}^{m} \left( y^{(i)} -  e^{\theta^T x^{(i)}} \right) x^{(i)} = X^T \left( Y -  e^{X\theta} \right)
\end{align*} 
$$

Given this new result, the batch gradient ascent formula becomes:
$$
\theta := \theta + \alpha X^T \left( Y -  e^{X\theta} \right)
$$
