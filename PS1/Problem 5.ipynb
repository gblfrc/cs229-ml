### (a) - i

Given the following vectors:

$$
X =
\begin{bmatrix}
- \ (x^{(1)})^T \ - \\
- \ (x^{(2)})^T \ - \\
\vdots \\
- \ (x^{(m)})^T \ - \\
\end{bmatrix}
\quad\quad
\theta =
\begin{bmatrix}
\theta_1 \\
\theta_2 \\
\vdots \\
\theta_m \\
\end{bmatrix}
\quad\quad
y =
\begin{bmatrix}
y^{(1)} \\
y^{(2)} \\
\vdots \\
y^{(m)} \\
\end{bmatrix}
$$

With the matrix-vector product, it is possible to compute:

$$
X \theta =
\begin{bmatrix}
(x^{(1)})^T \theta \\
(x^{(2)})^T \theta \\
\vdots \\
(x^{(m)})^T \theta \\
\end{bmatrix}
=
\begin{bmatrix}
\theta^T x^{(1)}  \\
\theta^T x^{(2)}  \\
\vdots \\
\theta^T x^{(m)}  \\
\end{bmatrix}
\implies
X \theta -y =
\begin{bmatrix}
\theta^T x^{(1)} - y^{(1)}  \\
\theta^T x^{(2)} - y^{(2)}  \\
\vdots \\
\theta^T x^{(m)} - y^{(m)} \\
\end{bmatrix}
$$

Therefore, $X \theta -y  \in \R^{m \times 1}$ with each element being $(X \theta -y)_i = \theta^T x^{(i)} - y^{(i)}$. Consequently, for the definition of transpose of a matrix, $(X \theta -y)^T  \in \R^{1 \times m}$. This implies that, in order for the product $(X \theta -y)^T W (X \theta -y)$ to make sense, $W$ must have suitable dimensions: $W \in \R^{m \times m}$. In that case, the product results in a scalar, which matches the expectations because $J(\theta)$ should be a scalar itself. At this point, it is necessary to determine the formulation of each element of the $W$ matrix.

The product in question resembles very closely a quadratic form, such as $z^T A z$. Therefore, its results can be written in a similar form:

$$
z^T A z = \sum_{j=1}^{m} \sum_{k=1}^{m} A_{jk} z_j z_k \implies (X \theta -y)^T W (X \theta -y) = \sum_{j=1}^{m} \sum_{k=1}^{m} W_{jk} (\theta^T x^{(j)} - y^{(j)}) (\theta^T x^{(k)} - y^{(k)})
$$

The summation just obtained includes also the "crossed" product of terms with different indices $j$ and $k$, whereas the target summation only includes terms associated to a single index. For the crossed terms to disappear from the summation, it is necessary for the associated $W_{jk}$ terms to be $0$. Therefore, $W$ must be a diagonal matrix. In such case, the summation would become:

$$
(X \theta -y)^T W (X \theta -y) = \sum_{i=1}^{m} W_{ii} (\theta^T x^{(i)} - y^{(i)}) (\theta^T x^{(i)} - y^{(i)}) = \sum_{i=1}^{m} W_{ii} (\theta^T x^{(i)} - y^{(i)})^2
$$

This formulation clearly suggests that the terms on the diagonal of $W$, should be the $w^{(i)}$ terms. Therefore:

$$
W =
\begin{bmatrix}
w^{(1)} & 0 & \cdots & 0 \\
0 & w^{(2)} & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & w^{(m)}
\end{bmatrix}
\implies
(X \theta -y)^T W (X \theta -y) = \sum_{i=1}^{m} w^{(i)} (\theta^T x^{(i)} - y^{(i)})^2
$$

which matches the initial $J(\theta)$ apart from the $\frac{1}{2}$ scaling term, which however doesn't affect the cost function optimization process. If the terms on the diagonal of $W$ were $\frac{1}{2}w^{(i)}$, the summations would match perfectly.


### (a) - ii

It was already demonstrated that, given a matrix $A \in \R^{n \times n}$  and the vectors $x, b \in \R^n$,
$$
\frac{\partial}{\partial x} \left[x^T A x \right] = 2Ax, \quad\quad \frac{\partial}{\partial x} \left[ b^T x \right] = b
$$

Leveraging the chain rule for derivatives:
$$
\nabla_\theta J(\theta) = \frac{\partial J(\theta)}{\partial (X\theta - y)} \frac{\partial (X\theta - y)}{\partial \theta} 
$$

We compute each of the 2 derivatives in isolation, starting from the derivative of $J(\theta)$. Assume $W$ has the form mentioned above, without including the $\frac{1}{2}$ in the terms on the diagonal. Computation is simplified by the first of the two derivation formulae presented above. 
$$
\begin{align*}
\frac{\partial J(\theta)}{\partial (X\theta - y)} &= \frac{\partial}{\partial (X\theta - y)}\left[\frac{1}{2}(X \theta -y)^T W (X \theta -y) \right] \\
&= \frac{1}{2} 2 W (X \theta -y) \\
&= W (X \theta -y)
\end{align*}
$$

Similarly, we can leverage the second derivation formula to compute the derivative of $(X\theta - y)$.

$$
\begin{align*}
\frac{\partial (X\theta - y)}{\partial \theta} = X^T
\end{align*}
$$

It follows that:
$$
\begin{align*}
\nabla_\theta J(\theta) &= X^T W (X \theta -y) \\
&= X^T W X \theta - X^T W y
\end{align*}
$$

where X^T was moved before the remaining terms to fitting the target dimensions.

The value of $\theta$ can be computed by setting to 0 the formula for the gradient above.

$$
\begin{align*}
X^T W X \theta - X^T W y &= 0 \\
X^T W X \theta &= X^T W y \\
\theta &= (X^T W X)^{-1} X^T W y \\
\end{align*}
$$



### (a) - iii

Consider the likelihood function:
$$
\begin{align*}
L(\theta) &= \prod_{i=1}^{m} p(y^{(i)}|x^{(i)};\theta) \\
&= \prod_{i=1}^{m} \frac{1}{\sqrt{2\pi}\sigma^{(i)}}\exp \left( -\frac{\left(y^{(i)}- \theta^T x^{(i)}\right)^2}{2\left(\sigma^{(i)}\right)^2} \right)
\end{align*}
$$

which becomes the log-likelihood:
$$
\begin{align*}
\ell(\theta) &= \sum_{i=1}^{m} \log \left[ \frac{1}{\sqrt{2\pi}\sigma^{(i)}}\exp \left( -\frac{\left(y^{(i)}- \theta^T x^{(i)}\right)^2}{2\left(\sigma^{(i)}\right)^2} \right) \right] \\
&= \sum_{i=1}^{m} \left[ \log \frac{1}{\sqrt{2\pi}\sigma^{(i)}} -\frac{\left(y^{(i)}- \theta^T x^{(i)}\right)^2}{2\left(\sigma^{(i)}\right)^2} \right]
\end{align*}
$$

To maximize the log-likelihood, it is necessary to compute its gradient and set it to $0$.


$$
\begin{align*}
\nabla_\theta \ell(\theta) &= \frac{\partial}{\partial\theta} \left[ \sum_{i=1}^{m} \left( \log \frac{1}{\sqrt{2\pi}\sigma^{(i)}} -\frac{\left(y^{(i)}- \theta^T x^{(i)}\right)^2}{2\left(\sigma^{(i)}\right)^2} \right) \right] \\
&=  \sum_{i=1}^{m} \frac{\partial}{\partial\theta} \left( \log \frac{1}{\sqrt{2\pi}\sigma^{(i)}} -\frac{\left(y^{(i)}- \theta^T x^{(i)}\right)^2}{2\left(\sigma^{(i)}\right)^2} \right) \\
&=  \sum_{i=1}^{m} \frac{\partial}{\partial\theta} \left(-\frac{\left(y^{(i)}- \theta^T x^{(i)}\right)^2}{2\left(\sigma^{(i)}\right)^2} \right) \\
&=  \sum_{i=1}^{m} -\frac{1}{2\left(\sigma^{(i)}\right)^2} \frac{\partial}{\partial\theta} \left(y^{(i)}- \theta^T x^{(i)}\right)^2  \\
&=  \sum_{i=1}^{m} -\frac{1}{2\left(\sigma^{(i)}\right)^2} \left[2\left(y^{(i)}- \theta^T x^{(i)}\right)x^{(i)}\right]  \\
&=  \sum_{i=1}^{m} -\frac{1}{\left(\sigma^{(i)}\right)^2}\left(y^{(i)}- \theta^T x^{(i)}\right)x^{(i)} \\
&=  \sum_{i=1}^{m} \frac{1}{\left(\sigma^{(i)}\right)^2}\left(\theta^T x^{(i)} - y^{(i)}\right)x^{(i)} \\
\end{align*}
$$

Assuming $w^{(i)}=\left(\sigma^{(i)}\right)^{-2}$, setting to $0$ the previous formula is equivalent to $X^T W (X \theta - y) = 0$, which is exactly the same result presented in the previous point.
