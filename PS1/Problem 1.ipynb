### (a)
Consider the empirical loss for logistic regression.
$$
J(\theta) = - \frac{1}{m} \sum_{i = 1}^{m} y^{(i)} \log \big( h_\theta (x^{(i)}) \big) + (1 - y^{(i)}) \log \big( 1 - h_\theta (x^{(i)}) \big)
$$,

The Hessian for such function can be computed as:
$$
\nabla_\theta^2J(\theta) = [ \nabla_\theta(\nabla_\theta(J(\theta)))_1 \quad \nabla_\theta(\nabla_\theta(J(\theta)))_2 \quad  \cdots \quad \nabla_\theta(\nabla_\theta(J(\theta)))_n ]
$$

which requires to compute first the gradient of $J(\theta)$ with respect to each component of $\theta$. By deriving such output, one can obtain each column of the target matrix. Therefore, consider the partial derivative of $J(\theta)$ on a generic component $\theta_j$ after replacing in the starting function $h_\theta(x^{(i)}) = g(\theta^T x^{(i)})$:

$$
\frac{\partial J(\theta)}{\partial \theta_j} = \frac{\partial }{\partial \theta_j} \left[ - \frac{1}{m} \sum_{i = 1}^{m} y^{(i)} \log \big( g(\theta^T x^{(i)}) \big) + (1 - y^{(i)}) \log \big( 1 - g(\theta^T x^{(i)}) \big) \right]
$$ 
$$
\frac{\partial J(\theta)}{\partial \theta_j} = - \frac{1}{m} \sum_{i = 1}^{m} y^{(i)} \left[ \frac{1}{g(\theta^T x^{(i)})} \left( g(\theta^T x^{(i)}) \right) \left(1- g(\theta^T x^{(i)})\right) x_j^{(i)} \right] - (1 - y^{(i)}) \left[ \frac{1}{\left(1- g(\theta^T x^{(i)})\right)} \left( g(\theta^T x^{(i)}) \right) \left(1- g(\theta^T x^{(i)})\right) x_j^{(i)} \right]
$$
$$
\frac{\partial J(\theta)}{\partial \theta_j} = - \frac{1}{m} \sum_{i = 1}^{m} y^{(i)} x_j^{(i)} \left(1- g(\theta^T x^{(i)})\right) - (1 - y^{(i)}) x_j^{(i)} \left( g(\theta^T x^{(i)}) \right) 
$$
$$
\frac{\partial J(\theta)}{\partial \theta_j} = - \frac{1}{m} \sum_{i = 1}^{m} y^{(i)} x_j^{(i)} - y^{(i)} x_j^{(i)} g(\theta^T x^{(i)}) -  x_j^{(i)} g(\theta^T x^{(i)}) + y^{(i)} x_j^{(i)} g(\theta^T x^{(i)})
$$
$$
\frac{\partial J(\theta)}{\partial \theta_j} = - \frac{1}{m} \sum_{i = 1}^{m} y^{(i)} x_j^{(i)} -  x_j^{(i)} g(\theta^T x^{(i)})
$$
$$
\frac{\partial J(\theta)}{\partial \theta_j} = \frac{1}{m} \sum_{i = 1}^{m} \left[ g(\theta^T x^{(i)}) - y^{(i)} \right] x_j^{(i)}
$$

which can be rewritten to matrix form as: $$\nabla_\theta J(\theta) = \frac{1}{m} X^T \left( g(X\theta) - Y \right)$$

By deriving this formula, one can compute each term of the Hessian:

$$
H_{jk} = \frac{\partial^2 J(\theta)}{\partial \theta_j \partial \theta_k} = \frac{\partial}{\partial \theta_k} \left[ \frac{1}{m} \sum_{i = 1}^{m} \left[ g(\theta^T x^{(i)}) - y^{(i)} \right] x_j^{(i)} \right]
$$
$$
H_{jk} =\frac{1}{m} \sum_{i = 1}^{m} g(\theta^T x^{(i)}) \left(1 - g(\theta^T x^{(i)})\right) x_j^{(i)} x_k^{(i)}
$$

which can be rewritten as matrix product: $$H_{jk} =\frac{1}{m} \left[ X^T g(X \theta) \left(1 - g(X \theta)\right) \right]  X$$

To demonstrate that the Hessian composed of such terms is positive semi-definite, one can demonstrate that: $$z^T H z \geq 0 \quad\quad \forall z \in \mathbb{R}^n$$

It was already shown that: $$z^T H z = \sum_{j=1}^{n} \sum_{k=1}^{n} H_{jk} z_j z_k$$

After replacing the form of $H_{jk}$ from before, one obtains: 
$$
z^T H z = \frac{1}{m} \sum_{j=1}^{n} \sum_{k=1}^{n} \sum_{i = 1}^{m} g(\theta^T x^{(i)}) \left(1 - g(\theta^T x^{(i)})\right) x_j^{(i)} x_k^{(i)} z_j z_k
$$
$$
z^T H z = \frac{1}{m} \sum_{i = 1}^{m} g(\theta^T x^{(i)}) \left(1 - g(\theta^T x^{(i)})\right) \left[ \sum_{j=1}^{n} \sum_{k=1}^{n} x_j^{(i)} x_k^{(i)} z_j z_k \right]
$$

It can be observed that: 
$$
\sum_{j=1}^{n} \sum_{k=1}^{n} x_j x_k z_j z_k = \left(x^T z \right)^2 \geq 0
$$

Therefore, the sign of the summation for $z^T H z$ depends only on the terms $g(\theta^T x^{(i)})$ and $\left(1 - g(\theta^T x^{(i)})\right)$. However, it can be shown that such terms are also always strictly positive. The first term is a sigmoid, which always has values between 0 and 1 (thus positive). The second term involves a sigmoid reflected on the $x$ axis, thus with values in $[-1, 0]$, shifted up by one on the $y$ axis, thus obtaining values in the $[0,1]$ range, i.e., positive. It follows that the summation is always positive and $H \succeq 0$.

### (d)

The target log-likelihood function is:

$$
\begin{aligned}
\ell(\phi, \mu_0, \mu_1, \Sigma)
&= \log \prod_{i=1}^{m} p(x^{(i)}, y^{(i)}; \phi, \mu_0, \mu_1, \Sigma) \\
&= \log \prod_{i=1}^{m} p(x^{(i)} \mid y^{(i)}; \mu_0, \mu_1, \Sigma)\ p(y^{(i)}; \phi)
\end{aligned}
$$

For the properties of logarithms, it can be expressed as:

$$
\begin{aligned}
\ell(\phi, \mu_0, \mu_1, \Sigma) &= \sum_{i=1}^{m} \log p(x^{(i)} \mid y^{(i)}; \mu_0, \mu_1, \Sigma)\, p(y^{(i)}; \phi) \\
&= \sum_{i=1}^{m} \log p(x^{(i)} \mid y^{(i)}; \mu_0, \mu_1, \Sigma) + \sum_{i=1}^{m} \log p(y^{(i)}; \phi)
\end{aligned}
$$

Expanding the probability terms:
$$
\begin{aligned}
\ell(\phi, \mu_0, \mu_1, \Sigma) &= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\}\log p(x^{(i)} \mid y^{(i)}; \mu_0, \Sigma) + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\}\log p(x^{(i)} \mid y^{(i)}; \mu_1, \Sigma) + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \log \phi + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \log (1-\phi) \\
&= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\}\log \frac{1}{(2\pi)^{d/2} |\Sigma|^{1/2}} \exp\left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) + 
\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\}\log \frac{1}{(2\pi)^{d/2} |\Sigma|^{1/2}} \exp\left( -\frac{1}{2} (x^{(i)} - \mu_1)^T \Sigma^{-1} (x^{(i)} - \mu_1) \right) + \\
& \quad + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \log \phi + 
\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \log (1-\phi) \\
&= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \log \frac{1}{(2\pi)^{d/2} |\Sigma|^{1/2}} + \\
& \quad + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \left( -\frac{1}{2} (x^{(i)} - \mu_1)^T \Sigma^{-1} (x^{(i)} - \mu_1) \right) + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \log \frac{1}{(2\pi)^{d/2} |\Sigma|^{1/2}} + \\
& \quad + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \log \phi +
\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \log (1-\phi) \\
&= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) +\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \left( -\frac{1}{2} (x^{(i)} - \mu_1)^T \Sigma^{-1} (x^{(i)} - \mu_1) \right) + \\
& \quad + \sum_{i=1}^{m} \log \frac{1}{(2\pi)^{d/2} |\Sigma|^{1/2}} + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \log \phi + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \log (1-\phi) \\
\end{aligned}
$$

For the final simplifications, focus on the following term:

$$
\sum_{i=1}^{m} \log \frac{1}{(2\pi)^{d/2} |\Sigma|^{1/2}} = - \sum_{i=1}^{m} \log (2\pi)^{d/2} |\Sigma|^{1/2} = - m \log (2\pi)^{d/2} |\Sigma|^{1/2} = - \frac{md}{2} \log (2\pi) - \frac{m}{2} \log |\Sigma|
$$

The formula for the maximum likelihood becomes:

$$
\begin{aligned}
\ell(\phi, \mu_0, \mu_1, \Sigma) &= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) +\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \left( -\frac{1}{2} (x^{(i)} - \mu_1)^T \Sigma^{-1} (x^{(i)} - \mu_1) \right) + \\
& \quad - \frac{md}{2} \log (2\pi) - \frac{m}{2} \log |\Sigma| + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \log \phi + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \log (1-\phi) \\
\end{aligned}
$$

To maximize the new expanded log-likelihood, one should compute its gradient with respect to each parameter and set all the computed gradients to zero. We start by computing the gradients.

$$
\begin{aligned}
\frac{\partial \ell}{\partial \phi} &= \frac{\partial}{\partial \phi} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \log \phi + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \log (1-\phi)\right] \\
&= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \frac{1}{\phi} - \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \frac{1}{(1-\phi)}  \\
&= \frac{1}{\phi} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} - \frac{1}{(1-\phi)} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\}
\end{aligned}
$$

$$
\begin{aligned}
\frac{\partial \ell}{\partial \mu_0} &= \frac{\partial}{\partial \mu_0} \left[ -\frac{1}{2} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right] \\
&=  -\frac{1}{2} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \ 2 \ \Sigma^{-1} (x^{(i)} - \mu_0) \\
&=  -\Sigma^{-1} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0)
\end{aligned}
$$

After deriving the product of vectors and $\Sigma$ as in the gradient of $x^T A x$, which equals $2Ax$. A similar formula can be obtained for the gradient of $\ell$ with respect to $\mu_1$:
$$
\frac{\partial \ell}{\partial \mu_1} = -\Sigma^{-1} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} (x^{(i)} - \mu_1)
$$

A more complex computation regards the gradient of $\Sigma$. 
$$
\begin{aligned}
\frac{\partial \ell}{\partial \Sigma} &= \frac{\partial}{\partial \Sigma} \left[ - \frac{m}{2} \log |\Sigma| + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) + \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \left( -\frac{1}{2} (x^{(i)} - \mu_1)^T \Sigma^{-1} (x^{(i)} - \mu_1) \right) \right] \\
\frac{\partial \ell}{\partial \Sigma} &= \frac{\partial}{\partial \Sigma} \left[ - \frac{m}{2} \log |\Sigma| \right] + \frac{\partial}{\partial \Sigma} \left[\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) \right] + \frac{\partial}{\partial \Sigma} \left[\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \left( -\frac{1}{2} (x^{(i)} - \mu_1)^T \Sigma^{-1} (x^{(i)} - \mu_1) \right) \right] \\
\end{aligned}
$$

For the first term, it can be shown that: 
$$
\frac{\partial \log |\Sigma|}{\partial \Sigma} = \Sigma^{-T} = \Sigma^{-1}
$$

where the last equality follows from the symmetry of $\Sigma$. Therefore:
$$
\frac{\partial}{\partial \Sigma} \left[ - \frac{m}{2} \log |\Sigma| \right] =  - \frac{m}{2} \Sigma^{-1}
$$

The remaining terms are similar, differing only for the parameter $\mu$ in each of them. Therefore, we will compute the gradient for one of them and extend the formula to the other one.
$$
\frac{\partial}{\partial \Sigma} \left[\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) \right] = \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \frac{\partial}{\partial \Sigma} \left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) \\
$$

In this term, the inverse of the target parameter appears. It can be shown that the derivative of an inverse matrix w.r.t. the base matrix is:
$$
\frac{\partial f(\Sigma^{-1})}{\partial \Sigma} = -\Sigma^{-1} \frac{\partial f(\Sigma^{-1})}{\partial \Sigma^{-1}} \Sigma^{-1}
$$

In this context, therefore: 

$$
\frac{\partial}{\partial \Sigma} \left( -\frac{1}{2} (x^{(i)} - \mu_0)^T \Sigma^{-1} (x^{(i)} - \mu_0) \right) = \frac{1}{2} \Sigma^{-1} (x^{(i)} - \mu_0) (x^{(i)} - \mu_0)^T\Sigma^{-1}
$$

which, re-introduced in the summation, becomes:
$$
\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} \left( -\frac{1}{2} \Sigma^{-1} (x^{(i)} - \mu_0) (x^{(i)} - \mu_0)^T\Sigma^{-1} \right) = \frac{1}{2} \Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0)(x^{(i)} - \mu_0)^T \right] \Sigma^{-1} 
$$

The final form, including all terms, of the gradient of the likelihood by $\Sigma$ is:
$$
\frac{\partial \ell}{\partial \Sigma} = 
- \frac{m}{2} \Sigma^{-1}
+\frac{1}{2} \Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0) (x^{(i)} - \mu_0)^T\right] \Sigma^{-1}
+\frac{1}{2} \Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} (x^{(i)} - \mu_1) (x^{(i)} - \mu_1)^T\right] \Sigma^{-1}  \\
$$

As said before, to estimate the parameter values, the computed gradients should be set to 0. Let's start with $\phi$:

$$
\begin{aligned}
\frac{\partial \ell}{\partial \phi} &= \frac{1}{\phi} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} - \frac{1}{(1-\phi)} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} = 0 \\
0 &= \frac{1}{\phi} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} - \frac{1}{(1-\phi)}  \left( m - \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} \right) \\
0 &= \left(\frac{1}{\phi} + \frac{1}{(1-\phi)}\right)  \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} - \frac{m}{(1-\phi)} \\
\frac{m}{(1-\phi)} &= \left(\frac{1}{\phi(1-\phi)}\right)  \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\}  \\
\phi &= \frac{1}{m}  \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\}  \\
\end{aligned}
$$

After that, we can compute the estimations for $\mu_0$ and $\mu_1$. Since they are much similar, we will estimate only $\mu_0$ and present the analogous formula for $\mu_1$. 

$$
\begin{aligned}
\frac{\partial \ell}{\partial \mu_0} &= -\Sigma^{-1} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0) = 0 \\
0 &= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0) \\
0 &= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} x^{(i)} - \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\}\mu_0 \\
\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\}\mu_0 &= \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} x^{(i)} \\
\mu_0 &= \frac{\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} x^{(i)}}{\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\}} \\
\end{aligned}
$$

Similarly:
$$
\mu_1 = \frac{\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} x^{(i)}}{\sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\}}
$$

Finally, from before:

$$
\frac{\partial \ell}{\partial \Sigma} = 
- \frac{m}{2} \Sigma^{-1}
+\frac{1}{2} \Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0)^T (x^{(i)} - \mu_0)\right] \Sigma^{-1}
+\frac{1}{2} \Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} (x^{(i)} - \mu_1)^T (x^{(i)} - \mu_1)\right] \Sigma^{-1} \\
$$

which, after being set to zero and pre-/post- multiplying all terms to remove halves and inverse matrices:

$$
0 = -m \Sigma^{-1} +
\Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0) (x^{(i)} - \mu_0)^T\right] \Sigma^{-1} +
\Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} (x^{(i)} - \mu_1) (x^{(i)} - \mu_1)^T\right] \Sigma^{-1}  \\
0 = -m +
\Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0) (x^{(i)} - \mu_0)^T\right] +
\Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} (x^{(i)} - \mu_1) (x^{(i)} - \mu_1)^T\right]  \\
m =
\Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0)(x^{(i)} - \mu_0)^T \right] +
\Sigma^{-1} \left[ \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} (x^{(i)} - \mu_1) (x^{(i)} - \mu_1)^T\right]  \\
\Sigma m =  \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0) (x^{(i)} - \mu_0)^T +  \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} (x^{(i)} - \mu_1) (x^{(i)} - \mu_1)^T \\
\Sigma = \frac{1}{m} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 0\} (x^{(i)} - \mu_0) (x^{(i)} - \mu_0)^T + \frac{1}{m} \sum_{i=1}^{m} \mathbf{1}\{y^{(i)} = 1\} (x^{(i)} - \mu_1) (x^{(i)} - \mu_1)^T 
$$
