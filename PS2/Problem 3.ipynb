# PS2 - Problem 3 - Bayesian Interpretation of Regularization

### (a)

From the definition of $\theta_{MAP}$, we know that:
$$
\theta_{MAP} = \arg \max_\theta p(\theta \mid x, y)
$$

By leveraging the chain rule for probability distributions, we can exapnd the joint probability $p(x,y,\theta)$ in two different ways:
$$
p(x,y,\theta) = p(\theta \mid x, y) \, p(x, y) \\
$$
$$
\begin{align*}
p(x,y,\theta) &= p(y \mid x, \theta) \, p(x, \theta) \\
&= p(y \mid x, \theta) \, p(\theta \mid x) \, p(x) \\
&= p(y \mid x, \theta) \, p(\theta) \, p(x)
\end{align*}
$$

where in the latter one we already applied the substitution $p(\theta \mid x) = p(\theta)$. The two quantities above must be equal, from which it follows

$$
\begin{align*}
p(\theta \mid x, y) \, p(x, y) &= p(y \mid x, \theta) \, p(\theta) \, p(x) \\
p(\theta \mid x, y) &= \frac{p(y \mid x, \theta) \, p(\theta) \, p(x)}{p(x, y)}  \\
p(\theta \mid x, y) &= p(y \mid x, \theta) \, p(\theta) \, \frac{p(x)}{p(x, y)}\\
\end{align*}
$$

The first term is exactly the probability that is being maximized to compute $\theta_{MAP}$. Therefore, we can replace it in the maximization with the second term:
$$
\begin{align*}
\theta_{MAP} = \arg \max_\theta p(\theta \mid x, y) &= \arg \max_\theta p(y \mid x, \theta) \, p(\theta) \, \frac{p(x)}{p(x, y)} \\
&= \arg \max_\theta p(y \mid x, \theta) \, p(\theta) \\
\end{align*}
$$

The final equality stems from the fact that, when maximizing in $\theta$, $\frac{p(x)}{p(x, y)}$ is just a scaling term and does not affect the mode.

### (b)

Recall that a generic distribution of a multivariate gaussian is:
$$
p(x) = \frac{1}{(2\pi)^{d/2} |\Sigma|^{1/2}} \exp\left(-\frac{1}{2} (x - \mu)^\top \Sigma^{-1} (x - \mu) \right)
$$

Since $\theta \sim \mathcal{N}(0, \eta^2 I)$, to obtain its distribution, we can replace in the previous formula the following quantities:
$$
\mu = 0, \quad\quad \Sigma = \eta^2I, \quad\quad |\Sigma| = \eta^{2d}, \quad\quad \Sigma^{-1} = \frac{1}{\eta^2} I
$$

Knowing that $\theta^T \theta = \|\theta\|_2^2$, we obtain:
$$
p(\theta) = \frac{1}{(2\pi)^{d/2} \eta^d} \exp\left(-\frac{\|\theta\|_2^2}{2\eta^2} \right)
$$

We can now replace such distribution in the final equation from the previous point.
$$
\begin{align*}
\theta_{MAP} &= \arg \max_\theta p(y \mid x, \theta) \, p(\theta) \\
&= \arg \max_\theta p(y \mid x, \theta) \, \frac{1}{(2\pi)^{d/2} \eta^d} \exp\left(-\frac{\|\theta\|_2^2}{2\eta^2} \right) \\
\end{align*}
$$

As already seen with the likelihood function, maximizing a function is equivalent to maximizing its logarithm, since the optimal point is not altered by the application of a monotonic function (such as the logarithm). Consequently, it holds:
$$
\begin{align*}
\theta_{MAP} &= \arg \max_\theta \log \left[ p(y \mid x, \theta) \, \frac{1}{(2\pi)^{d/2} \eta^d} \exp\left(-\frac{\|\theta\|_2^2}{2\eta^2} \right) \right] \\
&= \arg \max_\theta \left[ \log p(y \mid x, \theta) + \log \frac{1}{(2\pi)^{d/2} \eta^d} + \log \exp\left(-\frac{\|\theta\|_2^2}{2\eta^2} \right) \right] \\
&= \arg \max_\theta \left[ \log p(y \mid x, \theta) + \log \frac{1}{(2\pi)^{d/2} \eta^d} -\frac{1}{2\eta^2}\|\theta\|_2^2 \right] \\
\end{align*}
$$

In the final version of the objective to maximize, the second term does not depend on $\theta$, thus resulting in simply being a shifting term that does not affect the maximization. Therefore, it can be removed from the maximization objective. In addition, we could replace $\frac{1}{2\eta^2} = \lambda$, thus obtaining:
$$
\theta_{MAP} = \arg \max_\theta \left[ \log p(y \mid x, \theta) -\lambda\|\theta\|_2^2 \right] \\
$$

Since maximizing a function is equivalent to minimizing it negative version,
$$
\theta_{MAP} = \arg \min_\theta \left[ -\log p(y \mid x, \theta) + \lambda\|\theta\|_2^2 \right] \\
$$
which matches the demonstration target.

We also showed that the former formula holds for:
$$
\lambda = \frac{1}{2\eta^2}
$$

### (c)

From the problem description we understand that, for a single sample $(x^{(i)},  y^{(i)})$, it holds:
$$
y^{(i)} = \theta^T x^{(i)} + \epsilon^{(i)} \implies \epsilon^{(i)} = y^{(i)} - \theta^T x^{(i)}
$$

Since any $\epsilon^{(i)} \sim \mathcal{N}(0, \sigma^2)$, its probability distribution can be written as:
$$
p(\epsilon^{(i)})= \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(\epsilon^{(i)})^2}{2\sigma^2}\right)
$$

from which it follows that:
$$
p(y^{(i)} \mid x^{(i)}, \theta)= \frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{(y^{(i)} - \theta^T x^{(i)})^2}{2\sigma^2}\right)
$$

When accounting for samples from the whole dataset, the final equation from the previous point can be rewritten as:
$$
\theta_{MAP} = \arg \min_\theta \left[ -\log p(\vec{y} \mid X, \theta) + \lambda\|\theta\|_2^2 \right] \\
$$

We will call the argument of the minimization $J(\theta)$ and observe that it still contains a probability term which should be rewritten as a function of $\theta$.
$$
J(\theta) = -\log p(\vec{y} \mid X, \theta) + \lambda\|\theta\|_2^2
$$

The probability distribution above can be extended to account for the whole set of training samples. Assuming the samples are independent and identically distributed:
$$
\begin{align*}
p(\vec{y} \mid X, \theta) &= \prod_{i=1}^m p(y^{(i)} \mid x^{(i)}, \theta) \\
&= \prod_{i=1}^m \frac{1}{\sqrt{2\pi}\sigma} \exp\left(-\frac{(y^{(i)} - \theta^T x^{(i)})^2}{2\sigma^2}\right) \\
\end{align*}
$$

$J(\theta)$ contains, however, the logarithm of such distribution:
$$
\begin{align*}
\log p(\vec{y} \mid X, \theta) &= \log \left[ \prod_{i=1}^m p(y^{(i)} \mid x^{(i)}, \theta) \right]\\
&= \sum_{i=1}^m \left[\log \frac{1}{\sqrt{2\pi}\sigma} -\frac{(y^{(i)} - \theta^T x^{(i)})^2}{2\sigma^2} \right] \\
\end{align*}
$$

The just-computed logarithm can be replaced into the formula for $J(\theta)$. On top of this, since the first term in the summation does not depend on $\theta$, it would act as a simple shifting term in the function, thus not affecting the optimization procedure. Therefore, we can safely remove it straightaway. [The term would be deleted anyways when computing the gradient in $\theta$].

$$
J(\theta) = \sum_{i=1}^m \left[\frac{(y^{(i)} - \theta^T x^{(i)})^2}{2\sigma^2} \right] + \lambda\|\theta\|_2^2
$$

The summation in the function can be rewritten with a product of matrices and vectors, including the design matrix $X$ and the target vector $\vec{y}$.

$$
\begin{align*}
J(\theta) = \frac{1}{2\sigma^2} (\vec{y} - X\theta)^T(\vec{y} - X\theta) + \lambda\|\theta\|_2^2
\end{align*}
$$

To find the optimum value of this function, we should determine where its gradient is equal to $0$. We start by computing the gradient.
$$
\begin{align*}
\nabla_\theta J(\theta) &= \frac{\partial}{\partial\theta} \left[ \frac{1}{2\sigma^2} (\vec{y} - X\theta)^T(\vec{y} - X\theta) + \lambda\|\theta\|_2^2 \right] \\
&= \frac{1}{2\sigma^2} \frac{\partial}{\partial\theta} \left[ (\vec{y} - X\theta)^T(\vec{y} - X\theta)\right] + \frac{\partial}{\partial\theta} \left[\lambda\|\theta\|_2^2 \right]  \\
&= \frac{1}{2\sigma^2} 2X^T(X\theta-\vec{y}) + 2\lambda\theta  \\
&= \frac{1}{\sigma^2} X^T(X\theta-\vec{y}) + 2\lambda\theta  \\
\end{align*}
$$

Now we can set the gradient to $0$ and compute $\theta$ as a function of the other vectors. We also replace $\lambda$ with the value presented above to ease some simplifications.
$$
\nabla_\theta J(\theta) = \frac{1}{\sigma^2} X^T(X\theta-\vec{y}) + 2\frac{1}{2\eta^2}\theta = 0 \\
$$
$$
\begin{align*}
\frac{1}{\eta^2}\theta &= \frac{1}{\sigma^2} X^T(X\theta-\vec{y}) \\
\frac{1}{\eta^2}\theta &= \frac{1}{\sigma^2} X^TX\theta - \frac{1}{\sigma^2}X^T\vec{y} \\
X^TX\theta - \frac{\sigma^2}{\eta^2}\theta &= X^T\vec{y} \\
\left(X^TX - \frac{\sigma^2}{\eta^2}I\right)\theta &= X^T\vec{y} \\
\theta &= \left(X^TX - \frac{\sigma^2}{\eta^2}I\right)^{-1} X^T\vec{y} \\
\end{align*}
$$

The final equation provides an estimate for $\theta_{MAP}$.
$$
\theta_{MAP} = \left(X^TX - \frac{\sigma^2}{\eta^2}I\right)^{-1} X^T\vec{y}
$$

### (d)

From point (a) we know that $\theta_{MAP}$ can be found as:
$$
\theta_{MAP} = \arg \max_\theta p(y \mid x, \theta) \, p(\theta) \\
$$

To demonstrate the assertion for the current point, we follow a similar procedure to the one presented in the previous points. However, in this context, we start by rephrasing the maximization procedure by introducing logarithms and shifting to minimization.
$$
\begin{align*}
\theta_{MAP} &= \arg \max_\theta \log \left[ p(y \mid x, \theta) \, p(\theta) \right] \\
&= \arg \max_\theta \left[ \log p(y \mid x, \theta) + \log p(\theta) \right] \\
&= \arg \min_\theta \left[- \log p(y \mid x, \theta) - \log p(\theta) \right] \\
\end{align*}
$$

Similarly to the previous case, if we considered all the samples in the dataset, the expression could be rewritten as:
$$
\theta_{MAP} = \arg \min_\theta \left[- \log p(\vec{y} \mid X, \theta) - \log p(\theta) \right] \\
$$

We call the minimization objective $J(\theta)$.
$$
J(\theta) = - \log p(\vec{y} \mid X, \theta) - \log p(\theta)
$$

Since the linear regression model and the distribution of $\epsilon$ are the same as in the previous case, we directly replace the related term with the result from before. 
$$
J(\theta) = \frac{1}{2\sigma^2} (\vec{y} - X\theta)^T(\vec{y} - X\theta)  - \log p(\theta)
$$

We also know that
$$
(\vec{y} - X\theta)^T(\vec{y} - X\theta) = \|\vec{y} - X\theta\|_2^2
$$

which allows to rewrite $J(\theta)$ as:
$$
J(\theta) = \frac{1}{2\sigma^2} \|\vec{y} - X\theta\|_2^2  - \log p(\theta)
$$

In this formula, the only thing to further express is the final term with the logarithm of the prior distribution, which can be computed by recalling that $\theta \sim \mathcal{L}(0,bI)$. With a similar approach to the previous case, we obtain the prior distribution:
$$
p(\theta) = \frac{1}{(2b)^n} \exp\left(-\frac{\|\theta\|_1}{b}\right)
$$

whose logarithm is:
$$
\begin{align*}
\log p(\theta) &= \log \left[ \frac{1}{(2b)^n} \exp\left(-\frac{\|\theta\|_1}{b}\right) \right] \\
&= \log \frac{1}{(2b)^n} -\frac{\|\theta\|_1}{b}
\end{align*}
$$

In this final equality, the first term is not a function of $\theta$ and is irrelevant for the optimization procedure. Therefore, as done previously, we remove it straightaway and update $J(\theta)$:
$$
J(\theta) = \frac{1}{2\sigma^2} \|\vec{y} - X\theta\|_2^2 + \frac{1}{b} \|\theta\|_1
$$

Now, the $J(\theta)$ can be re-introduced in the minimization:
$$
\begin{align*}
\theta_{MAP} &= \arg \min_\theta \left[\frac{1}{2\sigma^2} \|\vec{y} - X\theta\|_2^2 + \frac{1}{b} \|\theta\|_1 \right] \\
&= \arg \min_\theta \left[\|\vec{y} - X\theta\|_2^2 + \frac{2\sigma^2}{b} \|\theta\|_1 \right] \\
\end{align*}
$$

where the last equality is obtained by multiplying all terms in the minimization objective by $2\sigma^2$. This is only a scaling term and does not affect the value of the optimal $\theta$. If we assumed
$$
\gamma = \frac{2\sigma^2}{b}
$$

the minimization objective would become
$$
J(\theta) = \|\vec{y} - X\theta\|_2^2 + \gamma \|\theta\|_1
$$

which matches the demonstration target.
