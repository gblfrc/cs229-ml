# PS2 - Problem 4 - Constructing kernels

### (a)

For a kernel function to be valid, the associated kernel matrix, given any finite set of points $\{x^{(i)}\}_{i=1}^m$, must be symmetric and positive semi-definite. The kernel matrix is defined as the matrix whose entry $K_{ij} = K(x^{(i)},x^{(j)})$. Therefore, it is symmetric if it holds that:
$$
K_{ij} = K_{ji} \Longleftrightarrow K(x^{(i)},x^{(j)}) = K(x^{(j)},x^{(i)})
$$

With respect to the kernels at hand, we verify
$$
K(x,z) = K_1(x,z) + K_2(x,z) = K_1(z,x) + K_2(z,x) = K(z,x)
$$

where the first and last equalities stem from the kernel definition, whereas the middle one stems from the simmetry of the given valid kernels $K_1$ and $K_2$. We observe, therefore, that also kernel $K$ is symmetric, and so will any of its kernel matrices be.

For the sake of notation simplicity, we observe that: 
$$
\begin{align*}
K(x^{(i)},x^{(j)}) &= K_1(x^{(i)},x^{(j)}) + K_2(x^{(i)},x^{(j)}) \\\
K_{ij} &= K_{1,ij} + K_{2,ij}
\end{align*}
$$
which implies that the kernel matrix of $K$ is equivalent to the sum of the kernel matrices of $K_1$ and $K_2$.

To demonstrate that $K$ is positive semi-definite, we compute its quadratic form.

$$
\begin{align*}
z^TKz &= \sum_{i=1}^m \sum_{j=1}^m z_i K_{ij} z_j \\
&= \sum_{i=1}^m \sum_{j=1}^m z_i (K_{1,ij} + K_{2,ij}) z_j \\
&= \sum_{i=1}^m \sum_{j=1}^m z_i K_{1,ij} z_j + \sum_{i=1}^m \sum_{j=1}^m z_i K_{2,ij} z_j \\
\end{align*}
$$

The quadratic form of $K$ is the sum of the quadratic forms of $K_1$ and $K_2$, which are valid kernels. Consequently, their quadratic forms are both greater or equal to zero, and their sum must be as well. It follows that $K$ must be a valid kernel as well.

### (b)

Assume we have the valid kernels $K_1$ and $K_2 = 2K_1$. It is easy to verify that $K_2$ is still a valid kernel: its kernel matrix is only a scaled version of that of $K_1$, thus still being symmetric, and its quadratic form is the quadratic form of $K_1$ multiplied by 2, thus being anyways greater or equal than 0.

However, given:
$$
K(x,z) = K_1(x,z) - K_2(x,z)
$$

the quadratic form of this kernel is:
$$
z^TKz = z^T(K_1 - K_2)z = z^T(K_1 - 2K_1)z = z^T(-K_1)z = -z^T(K_1)z
$$

Since $K_1$ is a valid kernel:
$$
z^T(K_1)z \geq 0 \implies -z^T(K_1)z \leq 0
$$

Therefore, the quadratic form of $K$ is negative semi-definite and the kernel is not valid. 

### (c)

The demonstration for this point is similar to that of point (a). We observe:
$$
K(x,z) = aK_1(x,z) = aK_1(z,x) = K(z,x)
$$

Therefore, the kernel matrix is symmetric. In addition:
$$
\begin{align*}
z^TKz &= \sum_{i=1}^m \sum_{j=1}^m z_i K_{ij} z_j \\
&= \sum_{i=1}^m \sum_{j=1}^m z_i a K_{1,ij} z_j \\
&= a \sum_{i=1}^m \sum_{j=1}^m z_i K_{1,ij} z_j
\end{align*}
$$

The summation is surely greater or equal to zero because it is the quadratic form of $K_1$, which is a valid kernel. Multiplying such value by a positve number does not change its sign and the output quadratic form is necessarily greater or equal to zero.


### (d)

The discussion is the same as the previous point, with the only difference that multiplying a positive quantity by a negative value makes it negative. Therefore $z^TKz \leq 0$ and the kernel is not valid.

### (e)

Following a similar reasoning as in point (a), we observe that:
$$
K(x,z) = K_1(x,z) \cdot K_2(x,z) = K_1(z,x) \cdot K_2(z,x) = K(z,x)
$$

The central equality stems from the fact that $K_1$ and $K_2$ are valid and, therefore, symmmetric kernels.

In addition, assume $x = x^{(i)}$ and $z = x^{(j)}$. Then we can write:
$$
K(x^{(i)},x^{(j)}) = K_{ij} = K_{1,ij} \cdot K_{2,ij} 
$$
which means that the target kernel matrix is the element-wise product of the two input kernel matrices.

We can compute the quadratic form of the kernel matrix of $K$:
$$
\begin{align*}
z^TKz &= \sum_{i=1}^m \sum_{j=1}^m z_i K_{ij} z_j \\
&= \sum_{i=1}^m \sum_{j=1}^m z_i K_{1,ij} K_{2,ij} z_j \\
\end{align*}
$$

Since $K_1$ and $K_2$ are valid kernels, there exist two feature maps $\phi_1$ and $\phi_2$ such that:
$$
K_{1,ij} = \phi_1(x^{(i)})^T\phi_1(x^{(j)}) = \sum_{h=1}^n \phi_1(x^{(i)})_h \phi_1(x^{(j)})_h\\
K_{2,ij} = \phi_2(x^{(i)})^T\phi_2(x^{(j)}) = \sum_{k=1}^n \phi_2(x^{(i)})_k \phi_2(x^{(j)})_k
$$

We can introduce these equalities in the quadratic form computation:
$$
\begin{align*}
z^TKz &= \sum_{i=1}^m \sum_{j=1}^m z_i z_j \sum_{h=1}^n \phi_1(x^{(i)})_h \phi_1(x^{(j)})_h \sum_{k=1}^n \phi_2(x^{(i)})_k \phi_2(x^{(j)})_k  \\
&= \sum_{h=1}^n \sum_{k=1}^n \sum_{i=1}^m \sum_{j=1}^m z_i \phi_2(x^{(i)})_k \phi_1(x^{(i)})_h \phi_1(x^{(j)})_h \phi_2(x^{(j)})_k z_j \\
&= \sum_{h=1}^n \sum_{k=1}^n \left( \sum_{i=1}^m \sum_{j=1}^m z_i \phi_1(x^{(i)})_h \phi_2(x^{(i)})_k \right)^2 \\
&\geq 0
\end{align*}
$$

Therefore, the kernel matrix is PSD and the kernel is valid.

### (f)

Similarly to the previous point, the kernel is symmetric:
$$
K(x,z) = f(x) \cdot f(z) = f(z) \cdot f(x) = K(x,z)
$$
where the central equality stems from the commutative property of products in $\R$. We can now compute the quadratic form of the kernel:
$$
\begin{align*}
z^TKz &= \sum_{i=1}^m \sum_{j=1}^m z_i K_{ij} z_j  \\
&= \sum_{i=1}^m \sum_{j=1}^m z_i f(x^{(i)}) f(x^{(j)}) z_j  \\
&= \left( \sum_{i=1}^m \sum_{j=1}^m z_i f(x^{(i)}) \right)^2 \\
&\geq 0
\end{align*}
$$

The kernel matrix is PSD and $K$ is a valid kernel.

### (g)

If $K_3$ is a valid kernel, then it represents an inner product of feature maps of its inputs and it is necessarily symmetric.
$$
K_3(a,b) = \psi(a)^T \psi(b) = \psi(b)^T \psi(a) = K_3(b,a)
$$

It follows that:
$$
K(x,z) = K_3(\phi(x),\phi(z)) = \psi(\phi(x))^T \psi(\phi(z)) = K_3(\phi(z),\phi(x)) = K(z,x)
$$

The quadratic form of the kernel is:
$$
\begin{align*}
z^TKz &= \sum_{i=1}^m \sum_{j=1}^m z_i K_{ij} z_j  \\
&= \sum_{i=1}^m \sum_{j=1}^m z_i  z_j \sum_{k=1}^n \psi(\phi(x^{(i)}))_k \psi(\phi(x^{(j)}))_k \\
&= \sum_{k=1}^n \sum_{i=1}^m \sum_{j=1}^m z_i  \psi(\phi(x^{(i)}))_k \psi(\phi(x^{(j)}))_k z_j \\
&= \sum_{k=1}^n \left( \sum_{i=1}^m \sum_{j=1}^m z_i \psi(\phi(x^{(i)}))_k \right)^2 \\
&\geq 0
\end{align*}
$$

The kernel matrix is PSD and the kernel is valid.

### (h)

The demonstration for the kernel symmetry proceeds as usual:
$$
K(x,z) = p(K_1(x,z)) = p(K_1(z,x)) = K(z,x)
$$
where the central equality is due to $K_1$ being a valid kernel.

We observe that, since $p$ is a polynomial with positive coefficients $a_i$ over $K_1$, it holds:
$$
p(K_{1,ij}) = \sum_{k=0}^q a_k K_{1,ij}^k = a_q K_{1,ij}^q + a_{q-1} K_{1,ij}^{q-1} + \cdots + a_1 K_{1,ij} + a_0
$$
where $q$ is the order of the polynomial. We can assume $q$ to be a positive even number without losing generality in the polynomial: if its degree was an odd number, we could still consider it as having a degree higher by 1 with an associated coefficient 0. This allows to rewrite the polynomials as:
$$
p(K_{1,ij}) = \sum_{k=0}^n \left[ a_{2k} K_{1,ij}^{2k} + a_{2k+1} K_{1,ij}^{2k+1} \right]
$$

where $n = q/2$. We now compute the quadratic form of $K$:
$$
\begin{align*}
z^TKz &= \sum_{i=1}^m \sum_{j=1}^m z_i K_{ij} z_j \\
&= \sum_{i=1}^m \sum_{j=1}^m z_i p(K_{1,ij}) z_j \\
&= \sum_{i=1}^m \sum_{j=1}^m z_i z_j \sum_{k=0}^n \left[ a_{2k} K_{1,ij}^{2k} + a_{2k+1} K_{1,ij}^{2k+1} \right] \\
&= \sum_{k=0}^n a_{2k} \sum_{i=1}^m \sum_{j=1}^m z_i z_j  K_{1,ij}^{2k} + \sum_{k=0}^n a_{2k+1} \sum_{i=1}^m \sum_{j=1}^m z_i z_j K_{1,ij}^{2k+1}\\
&= \sum_{k=0}^n a_{2k} K_{1,ij}^{2k} \sum_{i=1}^m \sum_{j=1}^m z_i z_j + \sum_{k=0}^n a_{2k+1} K_{1,ij}^{2k} \sum_{i=1}^m \sum_{j=1}^m z_i z_j K_{1,ij}\\
\end{align*}
$$

The final equation contains two terms of particular interest:
$$
\Sigma_z = \sum_{i=1}^m \sum_{j=1}^m z_i z_j = \left( \sum_{i=1}^m z_i \right)^2 \geq 0
$$
$$
\Sigma_K = \sum_{i=1}^m \sum_{j=1}^m z_i z_j K_{1,ij} \geq 0
$$

where the second one holds because the summation term is the quadratic form of $K_1$, which is a valid kernel. The quadratic form of $K$ becomes:
$$
z^TKz = \sum_{k=0}^n a_{2k} K_{1,ij}^{2k} \Sigma_z + \sum_{k=0}^n a_{2k+1} K_{1,ij}^{2k} \Sigma_K \geq 0
$$

because all terms in the summations are positive or equal to 0. Therefore, the kernel matrix of $K$ is PSD and $K$ is a valid kernel.
