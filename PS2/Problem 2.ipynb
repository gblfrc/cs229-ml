# PS2 - Problem 2 - Model Calibration

### (a)

Given the properties of the sigmoid function, all its outputs are strictly contained in the $(0,1)$ interval. This implies that, if $a = 0$ and $b = 0$, the index set coincides with the whole set of samples:
$$
I_{a,b} = \{x^{(i)},y^{(i)}\}_{i=1}^m
$$

Therefore, also its cardinality coincides with that of the overall set:
$$
\left|\{ i \in I_{a,b} \}\right| = m
$$

Following these observations, the target property can be rewritten as:
$$
\begin{align*}
\frac{\sum_{i \in I_{a,b}} P\left(y^{(i)} = 1 \mid x^{(i)}; \theta\right)}{\left|\{ i \in I_{a,b} \}\right|} 
&= \frac{\sum_{i \in I_{a,b}} \mathbb{I}\{y^{(i)} = 1\}}{\left|\{ i \in I_{a,b} \}\right|} \\
\frac{1}{m} \sum_{i=1}^{m} P\left(y^{(i)} = 1 \mid x^{(i)}; \theta\right) 
&= \frac{1}{m} \sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 1\}
\end{align*} 
$$

From the problem description, we know that:
$$
P\left(y^{(i)} = 1 \mid x^{(i)}; \theta\right) = h_\theta \left(x^{(i)}\right)
\implies
\frac{1}{m} \sum_{i=1}^{m} P\left(y^{(i)} = 1 \mid x^{(i)}; \theta\right) = \frac{1}{m} \sum_{i=1}^{m} h_\theta(x^{(i)})
$$

Additionally, since for the problem at hand $y^{(i)} \in \{0,1\}$, it holds that:
$$
\frac{1}{m} \sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 1\}
= 
\frac{1}{m} \sum_{i=1}^{m} y^{(i)}
$$

and the target property to show can be eventually rewritten as:
$$
\frac{1}{m} \sum_{i=1}^{m} h_\theta(x^{(i)}) = \frac{1}{m} \sum_{i=1}^{m} y^{(i)}
$$

It was already shown that the log-likelihood function for a logistic regression problem is:
$$
\ell(\theta) = \sum_{i=1}^{n} \left[ y^{(i)} \log h_\theta\big(x^{(i)}\big) + \left(1 - y^{(i)}\right) \log\left(1 - h_\theta\big(x^{(i)}\big)\right) \right]
$$

whose gradient is:
$$
\nabla_\theta \ell(\theta) = \sum_{i=1}^{n} \left( y^{(i)} - h_\theta\big(x^{(i)}\big)\right)x^{(i)}
$$

This relationship should hold for each element of the gradient, including the element associated to the bias term. Therefore:
$$
\nabla_\theta \ell(\theta)_k = \sum_{i=1}^{n} \left( y^{(i)} - h_\theta\big(x^{(i)}\big)\right)x^{(i)}_k \implies \nabla_\theta \ell(\theta)_0 = \sum_{i=1}^{n} \left( y^{(i)} - h_\theta\big(x^{(i)}\big)\right)x^{(i)}_0
$$

In previous problems, it was shown that the log-likelihood function is a convex function. Therefore, its optimum can be found when the gradient is $0$. This should hold for all the entries of the gradient. In addition, we recall that $x^{(i)}_0 = 1$ for all $x^{(i)}$ in our dataset. It follows:
$$
0 = \nabla_\theta \ell(\theta)_0 = \sum_{i=1}^{n} \left( y^{(i)} - h_\theta\big(x^{(i)}\big)\right)x^{(i)}_0 = \sum_{i=1}^{n} \left( y^{(i)} - h_\theta\big(x^{(i)}\big)\right) \cdot 1
$$
$$
\begin{align*}
\sum_{i=1}^{n} \left( y^{(i)} - h_\theta\big(x^{(i)}\big)\right) = 0 &\implies
\sum_{i=1}^{n} y^{(i)} - \sum_{i=1}^{n} h_\theta\big(x^{(i)}\big) = 0 \\
\sum_{i=1}^{n} y^{(i)} &= \sum_{i=1}^{n} h_\theta\big(x^{(i)}\big) \\
\end{align*}
$$

which is equivalent to the target property presented above.

### (b)

This question allows to demonstrate a few key observations:
- such a model does not exist for a logistic regression classifier
- yes, if the model is perfectly calibrated, it achieves perfect accuracy
- whether a model achieving perfect accuracy is well calibrated depends on the specific model and the definition of accuracy.

We start by assuming that a perfectly calibrated model exists for a binary classification task. Let's assume that the model outputs probabilities in $(0,1)$ and that we can define an interval $(a,b)$ so small that it includes a single sample. On this interval, the average of $y$ labels can only be either $0$ or $1$, whereas the average of the output probabilities is the probability of the sample itself, which necessarily differs from the average of the labels. Therefore, such interval must not exist and all outputs of a perfectly calibrated model must either be $0$'s or $1$'s. It follows that such model does not exist in a logistic regression setting, since the outputs of the sigmoid function can never be exactly $0$ or $1$. 

Nevertheless, let's assume that our binary classification model exists. Indeed, this could be a perceptron-like model assigning probability 1 to samples where $\theta^T x \geq 0$ and $0$ otherwise. In this case, let's assume that there exists a dataset such that the model does not achieve perfect accuracy on it:
$$
\{y^{(i)}\}_{i=1}^2 = \{0, 1\} \\
\{h_\theta(x^{(i)})\}_{i=1}^2 = \{1, 0\} \\
$$

This dataset is such that the model is calibrated on the interval $[0,1]$, where the average of both sets is 0.5, but it is not calibrated on any interval $[0,k)$, with $k \leq 1$. On these intervals, the average of of $y^{(i)}$ terms is 1, whereas the average of $h_\theta(x^{(i)})$ terms is 0, thus contradicting the initial assumption of a perfectly calibrated model. Therefore, if a model is perfectly calibrated, it also achieves perfect accuracy.

Regarding the opposite assertion, we already demonstrated that a perfectly calibrated model for a logistic regression scenario does not exist. Similar considerations can be made for other models returning values in $(0,1)$. 

For the perceptron-like model described above, instead, we now assume that it achieves perfect accuracy. This implies that all samples in the dataset are predicted with their correct label, i.e., either 0 or 1. We observe that there are no samples with $h_\theta(x^{(i)}) \in (0,1)$. Therefore, it makes no sense checking for calibration on intervals $(a,b) \subset (0,1)$. Similarly to the case above, we should check only intervals $[0,k)$, with $k \leq 1$, and intervals $(0,k]$, with $k \geq 0$. The first type of intervals contains only negative samples predicted as negatives, with matching averages between $y^{(i)}$ and $h_\theta(x^{(i)})$ terms. The same holds, on the second set of intervals, for positive samples. Therefore, the model is also perfectly calibrated.

### (c)

In the previous point (a) we have already presented the log-likelihood function as: 
$$
\ell(\theta) = \sum_{i=1}^{n} \left[ y^{(i)} \log h_\theta\big(x^{(i)}\big) + \left(1 - y^{(i)}\right) \log\left(1 - h_\theta\big(x^{(i)}\big)\right) \right]
$$

When adding $L_2$ regularization, the cost function becomes:
$$
\ell(\theta) = \sum_{i=1}^{n} \left[ y^{(i)} \log h_\theta\big(x^{(i)}\big) + \left(1 - y^{(i)}\right) \log\left(1 - h_\theta\big(x^{(i)}\big)\right) \right] - \frac{1}{2} \lambda ||\theta||_2^2
$$

Please notice that, in the previous function, the regularization term was subtracted from the likelihood instead of being added. The regularization term should prevent the norm of $\theta$ from exploding. Since the log-likelihood tends to be maximised, adding a positive norm of $\theta$ would result in increasing the cost function value, allowing to attain the maximum value in correspondence of an infinite norm of $\theta$. By subtracting the norm we obtain, instead, the desired constraining effect.

We proceed with a similar derivation as in the previous point (a), computing the gradient of the log likelihood with respect to $\theta$ and setting it to $0$.
$$
\nabla_\theta \ell(\theta) = \sum_{i=1}^{n} \left( y^{(i)} - h_\theta\big(x^{(i)}\big)\right)x^{(i)} - \lambda\theta = 0
$$

Focusing on the 0-component of the gradient vector, we obtain:
$$
\begin{align*}
\sum_{i=1}^{n} \left( y^{(i)} - h_\theta\big(x^{(i)}\big)\right)x^{(i)}_0 - \lambda\theta_0 &= 0 \\
\sum_{i=1}^{n} y^{(i)} - \sum_{i=1}^{n} h_\theta\big(x^{(i)}\big) - \lambda\theta_0 &= 0
\end{align*}
$$
$$
\sum_{i=1}^{n} h_\theta\big(x^{(i)}\big) + \lambda\theta_0 = \sum_{i=1}^{n} y^{(i)}
$$

Going back to the starting formula, we observe that the previous one is equivalent to:
$$
\sum_{i=1}^{n} P\left(y^{(i)} = 1 \mid x^{(i)}; \theta\right) + \lambda\theta_0 = \sum_{i=1}^{n} \mathbb{I}\{y^{(i)} = 1\}
$$

which shows that the model is no longer well calibrated. Instead, the model is calibrated by taking into account a shifting term which depends on the entity of the regularization.
