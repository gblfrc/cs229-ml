# PS2 - Problem 5 - Kernelizing the Perceptron

### (a) - i
The update rule presented in the problem description is:
$$
\begin{align*}
\theta^{(i+1)} &= \theta^{(i)} + \alpha \left(y^{(i+1)} - h_{\theta^{(i)}} \left(\phi \big(x^{(i+1)}\big)\right) \right) \phi \big(x^{(i+1)}\big)\\
&= \theta^{(i)} + \alpha \left(y^{(i+1)} - g \left( \big(\theta^{(i)}\big)^T\phi \big(x^{(i+1)}\big)\right) \right) \phi \big(x^{(i+1)}\big)\\
\end{align*}
$$

We start by exemplifying the update behavior for the initial iterations. Knowing that $\theta^{(0)} = \vec{0}$, we compute $\theta^{(1)}$ as:
$$
\begin{align*}
\theta^{(1)} &= \theta^{(0)} + \alpha \left(y^{(1)} - g \left( \big(\theta^{(0)}\big)^T\phi \big(x^{(1)}\big)\right) \right) \phi \big(x^{(1)}\big) \\
&= \vec{0} + \alpha \left(y^{(1)} - g \left(\vec{0}^T\phi \big(x^{(1)}\big)\right) \right) \phi \big(x^{(1)}\big) \\
&= \alpha \left(y^{(1)} - g \left(0\right) \right) \phi \big(x^{(1)}\big) \\
&= \alpha \left(y^{(1)} - 1 \right) \phi \big(x^{(1)}\big) \\
&= \beta_1 \phi \big(x^{(1)}\big)
\end{align*}
$$

Similarly
$$
\begin{align*}
\theta^{(2)} &= \theta^{(1)} + \alpha \left(y^{(2)} - g \left( \big(\theta^{(1)}\big)^T\phi \big(x^{(2)}\big)\right) \right) \phi \big(x^{(2)}\big) \\
&= \beta_1 \phi \big(x^{(1)}\big) + \alpha \left(y^{(2)} - g \left( \beta_1 \phi \big(x^{(1)}\big)^T\phi \big(x^{(2)}\big)\right) \right) \phi \big(x^{(2)}\big) \\
&= \beta_1 \phi \big(x^{(1)}\big) + \beta_2 \phi \big(x^{(2)}\big) \\
&= \sum_{j=1}^2 \beta_j \phi \big(x^{(j)}\big)\\\\

\theta^{(3)} &= \theta^{(2)} + \alpha \left(y^{(3)} - g \left( \big(\theta^{(2)}\big)^T\phi \big(x^{(3)}\big)\right) \right) \phi \big(x^{(3)}\big) \\
&= \sum_{j=1}^2 \beta_j \phi \big(x^{(j)}\big) + \beta_3 \phi \big(x^{(3)}\big) \\
&= \sum_{j=1}^3 \beta_j \phi \big(x^{(j)}\big) \\
\end{align*}
$$

It follows that the value of $\theta_i$ can be generalized as:
$$
\theta^{(i)} = \sum_{j=1}^i \beta_j \phi \big(x^{(j)}\big) \\
$$

### (a) - ii

Let's assume that $\theta^{(i)}$ is the final value of $\theta$ after the training has completed. In such case, we could use it for predicting on a new sample $z$. The prediction would be:

For prediction:
$$
\begin{align*}
h_\theta(z) &= g \left(\big(\theta^{(i)}\big)^T \phi(z) \right) \\
&= g \left(\sum_{j=1}^i \beta_j \phi \big(x^{(j)}\big)^T \phi(z) \right) \\
&= sign \left(\sum_{j=1}^i \beta_j K \left( x^{(j)}, z \right) \right) \\
\end{align*}
$$

### (a) - iii

The update formula for $\theta^{(i+1)}$ can now be rewritten as:
$$
\begin{align*}
\theta^{(i+1)} &= \theta^{(i)} + \alpha \left(y^{(i+1)} - g \left( \big(\theta^{(i)}\big)^T\phi \big(x^{(i+1)}\big)\right) \right) \phi \big(x^{(i+1)}\big)\\
&= \sum_{j=1}^i \beta_j \phi \big(x^{(j)}\big) + \alpha \left(y^{(i+1)} - g \left( \sum_{j=1}^i \beta_j \phi \big(x^{(j)}\big)^T\phi \big(x^{(i+1)}\big)\right) \right) \phi \big(x^{(i+1)}\big)\\
\end{align*}
$$

In a previous point, we have already observed that $\beta_{i+1}$ is the coefficient of the $\phi \big(x^{(i+1)}\big)$ term. Therefore:
$$
\begin{align*}
\beta_{i+1} &= \alpha \left(y^{(i+1)} - g \left( \sum_{j=1}^i \beta_j \phi \big(x^{(j)}\big)^T \phi \big(x^{(i+1)}\big)\right) \right) \\
&= \alpha \left(y^{(i+1)} - sign \left( \sum_{j=1}^i \beta_j K \left( x^{(j)}, x^{(i+1)}\right) \right) \right) \\
\end{align*}
$$
