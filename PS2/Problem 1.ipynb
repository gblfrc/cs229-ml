By plotting the data in both datasets (see below), we observe that samples of A are not linearly separable, whereas samples of B are. In addition, after updating the code to print the values of theta at each iteration, we observe that its norm continuously increases when training on B. When triaining on A, as already said, the vector converges after 30,386 iterations.

From the code, it is possible to infer that the gradient of the cost function is computed as:
$$
\begin{align*}
\nabla_\theta J(\theta) &= -\frac{1}{m} \sum_{i=1}^m \frac{y^{(i)}}{1+\exp(y^{(i)} \theta^T x^{(i)})} x^{(i)} \\
&= -\frac{1}{m} \sum_{i=1}^m \frac{y^{(i)}}{1+\exp(y^{(i)} \theta^T x^{(i)})} \frac{\exp(-y^{(i)} \theta^T x^{(i)})}{\exp(-y^{(i)} \theta^T x^{(i)})} x^{(i)} \\
&= -\frac{1}{m} \sum_{i=1}^m \frac{\exp(-y^{(i)} \theta^T x^{(i)})}{1+\exp(-y^{(i)} \theta^T x^{(i)})} y^{(i)}  x^{(i)} \\
\end{align*}
$$

It is possible to observe that such expression matches the gradient of the cost function:
$$
J(\theta) = \frac{1}{m} \sum_{i=1}^m \log\left(1+\exp\left(-y^{(i)} \theta^T x^{(i)}\right)\right)
$$

In this formula, the exponent of the exponential term matches the functional margin defined in the context of SVMs. One could observe that the value of the cost function decreases as the absolute value of the exponent increases, which for linearly separable data can be achieved by simply increasing $\theta$ over and over. For this reason, when training on B, the value of $\theta$ steadily increases. When training on A, on the other hand, the value of $\theta$ is constrained by the presence of mixed positive and negative element (non-linearly-separable) and the computation of $\theta$ converges.

No, SVMs introduce by definition a constraint on the norm of $\theta$, thus preventing phenomena such as the one observed by training a logistic regressor on dataset B. SVMs are designed to maximize the geometric margin, which is a normalized version of the functional margin.

In addition, the hinge loss has the form:
$$
J(\theta) = \max(0,1-y \cdot \hat{y}), \quad with \, \hat{y} = \theta^T x.
$$

If the sets were linearly separable, then for every sample the second term would be lower than 0 and the whole loss would be 0. Therefore, continuously increasing $\theta$ would not result in steadily lowering the loss, and the phenomenon observed before would be avoided.
